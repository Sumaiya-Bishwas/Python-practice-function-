```python
# ==============================
# Python Functions Example
# ==============================

# Function to display a welcome message
def welcome(name):
    print(f"\nHello, {name}! Welcome to the Function Demo.\n")


# Function to calculate factorial using recursion
def factorial(n):
    if n == 0 or n == 1:
        return 1
    return n * factorial(n - 1)


# Function to check if a number is prime
def is_prime(num):
    if num < 2:
        return False
    for i in range(2, int(num ** 0.5) + 1):
        if num % i == 0:
            return False
    return True


# Function to calculate average
def average(numbers):
    return sum(numbers) / len(numbers)


# Main function
def main():
    name = input("Enter your name: ")
    welcome(name)

    number = int(input("Enter a number: "))

    print(f"\nFactorial of {number} = {factorial(number)}")

    if is_prime(number):
        print(f"{number} is a Prime Number.")
    else:
        print(f"{number} is NOT a Prime Number.")

    values = list(map(int, input("\nEnter numbers separated by spaces: ").split()))

    print(f"Average = {average(values):.2f}")


# Program starts here
if __name__ == "__main__":
    main()
```

```python
def fibonacci():
    cache = {}

    def fib(n):
        if n in cache:
            return cache[n]

        if n <= 1:
            return n

        cache[n] = fib(n - 1) + fib(n - 2)
        return cache[n]

    return fib

fib = fibonacci()

for i in range(20):
    print(fib(i), end=" ")
```

```python
import re

def validate_password(password):
    rules = {
        "Minimum Length": len(password) >= 8,
        "Uppercase": bool(re.search(r"[A-Z]", password)),
        "Lowercase": bool(re.search(r"[a-z]", password)),
        "Digit": bool(re.search(r"\d", password)),
        "Special Character": bool(re.search(r"[!@#$%^&*()_+=-]", password))
    }

    for rule, status in rules.items():
        print(f"{rule}: {'✔' if status else '✘'}")

    return all(rules.values())

password = input("Enter Password: ")

if validate_password(password):
    print("Strong Password")
else:
    print("Weak Password")
```

```python
import random
import time

def retry(attempts=3):
    def decorator(func):
        def wrapper(*args, **kwargs):
            for attempt in range(1, attempts + 1):
                try:
                    return func(*args, **kwargs)
                except Exception as e:
                    print(f"Attempt {attempt} Failed: {e}")
                    time.sleep(1)

            print("All attempts failed.")
        return wrapper
    return decorator

@retry(5)
def unstable():
    if random.randint(1, 4) != 1:
        raise Exception("Random Error")
    return "Success"

print(unstable())
```

```python
def word_frequency(sentence):

    words = sentence.lower().split()

    frequency = {}

    for word in words:
        frequency[word] = frequency.get(word, 0) + 1

    return frequency


text = input("Enter a sentence: ")

result = word_frequency(text)

print("\nWord Frequency:")

for word, count in result.items():
    print(word, ":", count)
```

```python
def encrypt(text, shift):
    encrypted = ""

    for char in text:
        if char.isalpha():
            start = 65 if char.isupper() else 97
            encrypted += chr((ord(char) - start + shift) % 26 + start)
        else:
            encrypted += char

    return encrypted


message = input("Enter Message: ")
shift = int(input("Enter Shift Value: "))

print("Encrypted Message:", encrypt(message, shift))
```

```python
def is_palindrome(sentence):
    cleaned = ""

    for char in sentence.lower():
        if char.isalnum():
            cleaned += char

    return cleaned == cleaned[::-1]


text = input("Enter a sentence: ")

if is_palindrome(text):
    print("Palindrome")
else:
    print("Not Palindrome")
```

```python
def password_score(password):

    score = 0

    if len(password) >= 8:
        score += 1

    if any(c.isupper() for c in password):
        score += 1

    if any(c.islower() for c in password):
        score += 1

    if any(c.isdigit() for c in password):
        score += 1

    if any(not c.isalnum() for c in password):
        score += 1

    levels = ["Very Weak", "Weak", "Average", "Good", "Strong", "Excellent"]

    return levels[score]


password = input("Enter Password: ")

print("Strength:", password_score(password))
```

```python
expenses = []


def add_expense():
    category = input("Enter Category: ")
    amount = float(input("Enter Amount: "))

    expenses.append({
        "Category": category,
        "Amount": amount
    })

    print("Expense Added Successfully!\n")


def view_expenses():
    if not expenses:
        print("No Expenses Found!\n")
        return

    print("\n----- Expense List -----")
    total = 0

    for i, expense in enumerate(expenses, start=1):
        print(f"{i}. {expense['Category']} - ${expense['Amount']:.2f}")
        total += expense["Amount"]

    print(f"\nTotal Expense: ${total:.2f}\n")


def highest_expense():
    if not expenses:
        print("No Expenses Found!\n")
        return

    highest = max(expenses, key=lambda x: x["Amount"])

    print("\nHighest Expense")
    print("Category:", highest["Category"])
    print("Amount: $", highest["Amount"])


def search_category():
    category = input("Enter Category to Search: ").lower()

    found = False

    for expense in expenses:
        if expense["Category"].lower() == category:
            print(f"{expense['Category']} - ${expense['Amount']:.2f}")
            found = True

    if not found:
        print("Category Not Found.")


def menu():
    while True:
        print("\n====== Expense Tracker ======")
        print("1. Add Expense")
        print("2. View Expenses")
        print("3. Highest Expense")
        print("4. Search by Category")
        print("5. Exit")

        choice = input("Enter Choice: ")

        if choice == "1":
            add_expense()

        elif choice == "2":
            view_expenses()

        elif choice == "3":
            highest_expense()

        elif choice == "4":
            search_category()

        elif choice == "5":
            print("Goodbye!")
            break

        else:
            print("Invalid Choice!")


menu()
```

```python
books = []


def add_book():
    title = input("Enter Book Title: ")
    author = input("Enter Author Name: ")

    books.append({
        "Title": title,
        "Author": author,
        "Available": True
    })

    print("Book Added Successfully!\n")


def view_books():
    if not books:
        print("No Books Available.\n")
        return

    print("\n----- Library Books -----")

    for i, book in enumerate(books, start=1):
        status = "Available" if book["Available"] else "Borrowed"
        print(f"{i}. {book['Title']} | {book['Author']} | {status}")


def borrow_book():
    title = input("Enter Book Title: ")

    for book in books:
        if book["Title"].lower() == title.lower():
            if book["Available"]:
                book["Available"] = False
                print("Book Borrowed Successfully!\n")
            else:
                print("Book is Already Borrowed.\n")
            return

    print("Book Not Found.\n")


def return_book():
    title = input("Enter Book Title: ")

    for book in books:
        if book["Title"].lower() == title.lower():
            if not book["Available"]:
                book["Available"] = True
                print("Book Returned Successfully!\n")
            else:
                print("Book is Already Available.\n")
            return

    print("Book Not Found.\n")


def menu():
    while True:
        print("\n====== Library Management ======")
        print("1. Add Book")
        print("2. View Books")
        print("3. Borrow Book")
        print("4. Return Book")
        print("5. Exit")

        choice = input("Enter Choice: ")

        if choice == "1":
            add_book()
        elif choice == "2":
            view_books()
        elif choice == "3":
            borrow_book()
        elif choice == "4":
            return_book()
        elif choice == "5":
            print("Thank You!")
            break
        else:
            print("Invalid Choice!\n")


menu()
```

```python
import time

def execution_time(func):
    def wrapper(*args, **kwargs):
        start = time.perf_counter()
        result = func(*args, **kwargs)
        end = time.perf_counter()

        print(f"\nFunction: {func.__name__}")
        print(f"Execution Time: {end - start:.6f} seconds")
        return result

    return wrapper


@execution_time
def find_prime_numbers(limit):
    primes = []

    for num in range(2, limit + 1):
        is_prime = True

        for i in range(2, int(num ** 0.5) + 1):
            if num % i == 0:
                is_prime = False
                break

        if is_prime:
            primes.append(num)

    return primes


@execution_time
def calculate_sum(limit):
    total = sum(range(1, limit + 1))
    return total


prime_list = find_prime_numbers(5000)
print(f"Total Prime Numbers: {len(prime_list)}")

total = calculate_sum(1_000_000)
print(f"Sum = {total}")
```

```python
import random
import time


def retry(max_attempts=3, delay=1):
    def decorator(func):
        def wrapper(*args, **kwargs):
            for attempt in range(1, max_attempts + 1):
                try:
                    result = func(*args, **kwargs)
                    print(f"Success on attempt {attempt}")
                    return result
                except Exception as error:
                    print(f"Attempt {attempt} failed: {error}")

                    if attempt < max_attempts:
                        print(f"Retrying in {delay} second(s)...\n")
                        time.sleep(delay)
                    else:
                        print("Maximum retry attempts reached.")
                        return None

        return wrapper
    return decorator


@retry(max_attempts=5, delay=2)
def connect_to_server():
    if random.choice([True, False]):
        raise ConnectionError("Server is unreachable!")

    return "Connected Successfully."


result = connect_to_server()

if result:
    print(result)
```

```python
def memoize(func):
    cache = {}

    def wrapper(*args):
        if args in cache:
            print(f"Cache Hit -> {args}")
            return cache[args]

        print(f"Calculating -> {args}")
        result = func(*args)
        cache[args] = result
        return result

    return wrapper


@memoize
def fibonacci(n):
    if n <= 1:
        return n
    return fibonacci(n - 1) + fibonacci(n - 2)


number = 35

print(f"Fibonacci({number}) =", fibonacci(number))
print(f"Fibonacci({number}) =", fibonacci(number))
```

```python
def scheduler():
    tasks = []

    def add_task(name, priority):
        tasks.append((priority, name))
        tasks.sort(reverse=True)

    def show_tasks():
        print("\nTask List")
        print("-" * 25)

        if not tasks:
            print("No tasks available.")
            return

        for priority, name in tasks:
            print(f"Priority {priority} -> {name}")

    def complete_task():
        if tasks:
            priority, name = tasks.pop(0)
            print(f"Completed: {name}")
        else:
            print("No task to complete.")

    return add_task, show_tasks, complete_task


add, show, complete = scheduler()

add("Finish Python Assignment", 2)
add("Prepare Presentation", 5)
add("Practice DSA", 4)
add("Upload GitHub Project", 3)

show()

complete()
complete()

show()
```

```python
def pipeline(*functions):

    def execute(data):
        for func in functions:
            data = func(data)
        return data

    return execute


def remove_spaces(text):
    return text.strip()


def to_upper(text):
    return text.upper()


def replace_spaces(text):
    return text.replace(" ", "_")


def add_extension(text):
    return text + ".txt"


process = pipeline(
    remove_spaces,
    to_upper,
    replace_spaces,
    add_extension
)

filename = process("   python function project   ")

print(filename)
```

```python
from datetime import datetime


def call_logger(func):
    history = []

    def wrapper(*args, **kwargs):
        result = func(*args, **kwargs)

        history.append({
            "time": datetime.now().strftime("%H:%M:%S"),
            "arguments": args,
            "result": result
        })

        return result

    wrapper.history = history
    return wrapper


@call_logger
def cube(number):
    return number ** 3


cube(2)
cube(5)
cube(10)

print("\nCall History")
print("-" * 35)

for item in cube.history:
    print(item)
```

```python
def count_calls(func):
    def wrapper(*args, **kwargs):
        wrapper.calls += 1
        print(f"{func.__name__} called {wrapper.calls} time(s)")
        return func(*args, **kwargs)

    wrapper.calls = 0
    return wrapper


@count_calls
def greet(name):
    print(f"Hello, {name}!")


greet("Alice")
greet("Bob")
greet("Charlie")
greet("David")

print("Total Calls:", greet.calls)
```

```python
def text_editor():
    text = ""
    history = []

    def write(new_text):
        nonlocal text
        history.append(text)
        text += new_text

    def undo():
        nonlocal text
        if history:
            text = history.pop()

    def display():
        print("Current Text:", text)

    return write, undo, display


write, undo, display = text_editor()

write("Python ")
write("Functions ")
write("are ")
write("Powerful!")

display()

undo()
display()

undo()
display()
```

```python
def transaction_manager(initial_balance):
    balance = initial_balance
    transactions = []

    def deposit(amount):
        nonlocal balance
        if amount > 0:
            balance += amount
            transactions.append(f"Deposited: {amount}")

    def withdraw(amount):
        nonlocal balance
        if amount <= balance:
            balance -= amount
            transactions.append(f"Withdrawn: {amount}")
        else:
            transactions.append("Withdrawal Failed")

    def statement():
        print("\nTransaction History")
        print("-" * 30)

        for item in transactions:
            print(item)

        print(f"\nCurrent Balance: {balance}")

    return deposit, withdraw, statement


deposit, withdraw, statement = transaction_manager(10000)

deposit(2500)
withdraw(1800)
withdraw(15000)
deposit(700)

statement()
```

```python
def add_book(library):
    print("\nAdd New Book")

    book_id = input("Book ID: ")

    for book in library:
        if book["id"] == book_id:
            print("Book ID already exists.")
            return

    title = input("Book Title: ")
    author = input("Author Name: ")

    while True:
        try:
            quantity = int(input("Quantity: "))
            if quantity >= 0:
                break
            else:
                print("Quantity cannot be negative.")
        except ValueError:
            print("Enter a valid number.")

    library.append({
        "id": book_id,
        "title": title,
        "author": author,
        "quantity": quantity
    })

    print("Book added successfully.\n")


def view_books(library):

    if len(library) == 0:
        print("\nNo books available.\n")
        return

    print("\nAvailable Books")
    print("-" * 60)

    for book in library:
        print(f"ID       : {book['id']}")
        print(f"Title    : {book['title']}")
        print(f"Author   : {book['author']}")
        print(f"Quantity : {book['quantity']}")
        print("-" * 60)


def search_book(library):

    keyword = input("Enter book title: ").lower()

    found = False

    for book in library:
        if keyword in book["title"].lower():
            print("\nBook Found")
            print(book)
            found = True

    if not found:
        print("Book not found.")


def borrow_book(library):

    book_id = input("Enter Book ID: ")

    for book in library:
        if book["id"] == book_id:

            if book["quantity"] > 0:
                book["quantity"] -= 1
                print("Book borrowed successfully.")
            else:
                print("Book is out of stock.")
            return

    print("Book not found.")


def return_book(library):

    book_id = input("Enter Book ID: ")

    for book in library:
        if book["id"] == book_id:
            book["quantity"] += 1
            print("Book returned successfully.")
            return

    print("Book not found.")


def update_book(library):

    book_id = input("Enter Book ID: ")

    for book in library:

        if book["id"] == book_id:

            print("\nLeave blank if you don't want to change.")

            new_title = input("New Title: ")
            new_author = input("New Author: ")
            new_quantity = input("New Quantity: ")

            if new_title != "":
                book["title"] = new_title

            if new_author != "":
                book["author"] = new_author

            if new_quantity != "":
                book["quantity"] = int(new_quantity)

            print("Book updated successfully.")
            return

    print("Book not found.")


def remove_book(library):

    book_id = input("Enter Book ID: ")

    for book in library:

        if book["id"] == book_id:
            library.remove(book)
            print("Book removed successfully.")
            return

    print("Book not found.")


def total_books(library):

    total = 0

    for book in library:
        total += book["quantity"]

    print("\nTotal Different Books :", len(library))
    print("Total Available Copies :", total)


def menu():

    library = []

    while True:

        print("""
========== Library Management ==========
1. Add Book
2. View Books
3. Search Book
4. Borrow Book
5. Return Book
6. Update Book
7. Remove Book
8. Total Books
9. Exit
========================================
""")

        choice = input("Enter your choice: ")

        if choice == "1":
            add_book(library)

        elif choice == "2":
            view_books(library)

        elif choice == "3":
            search_book(library)

        elif choice == "4":
            borrow_book(library)

        elif choice == "5":
            return_book(library)

        elif choice == "6":
            update_book(library)

        elif choice == "7":
            remove_book(library)

        elif choice == "8":
            total_books(library)

        elif choice == "9":
            print("Thank you for using the Library Management System.")
            break

        else:
            print("Invalid choice. Try again.")


def main():
    menu()


if __name__ == "__main__":
    main()
```

```python
def add_habit(habits):
    name = input("Enter habit name: ")

    for habit in habits:
        if habit["name"].lower() == name.lower():
            print("Habit already exists.\n")
            return

    habits.append({
        "name": name,
        "completed": 0,
        "missed": 0
    })

    print("Habit added successfully.\n")


def complete_habit(habits):
    name = input("Habit completed: ")

    for habit in habits:
        if habit["name"].lower() == name.lower():
            habit["completed"] += 1
            print("Great job! Habit marked as completed.\n")
            return

    print("Habit not found.\n")


def miss_habit(habits):
    name = input("Habit missed: ")

    for habit in habits:
        if habit["name"].lower() == name.lower():
            habit["missed"] += 1
            print("Habit marked as missed.\n")
            return

    print("Habit not found.\n")


def show_report(habits):

    if len(habits) == 0:
        print("No habits found.\n")
        return

    print("\n========== HABIT REPORT ==========")

    for habit in habits:

        total = habit["completed"] + habit["missed"]

        if total == 0:
            success = 0
        else:
            success = (habit["completed"] / total) * 100

        print("----------------------------")
        print("Habit      :", habit["name"])
        print("Completed  :", habit["completed"])
        print("Missed     :", habit["missed"])
        print("Success %  :", round(success, 2))

    print("----------------------------\n")


def best_habit(habits):

    if len(habits) == 0:
        print("No habits available.\n")
        return

    best = max(habits, key=lambda x: x["completed"])

    print("\nBest Performing Habit")
    print("Habit :", best["name"])
    print("Completed :", best["completed"])
    print()


def remove_habit(habits):

    name = input("Enter habit name to remove: ")

    for habit in habits:
        if habit["name"].lower() == name.lower():
            habits.remove(habit)
            print("Habit removed.\n")
            return

    print("Habit not found.\n")


def menu():

    habits = []

    while True:

        print("""
========= HABIT TRACKER =========
1. Add Habit
2. Complete Habit
3. Miss Habit
4. Show Report
5. Best Habit
6. Remove Habit
7. Exit
=================================
""")

        choice = input("Choose option: ")

        if choice == "1":
            add_habit(habits)

        elif choice == "2":
            complete_habit(habits)

        elif choice == "3":
            miss_habit(habits)

        elif choice == "4":
            show_report(habits)

        elif choice == "5":
            best_habit(habits)

        elif choice == "6":
            remove_habit(habits)

        elif choice == "7":
            print("Program closed.")
            break

        else:
            print("Invalid choice.\n")


def main():
    menu()


main()
```

```python
def check_password(password):

    score = 0

    if len(password) >= 8:
        score += 1

    if any(ch.isupper() for ch in password):
        score += 1

    if any(ch.islower() for ch in password):
        score += 1

    if any(ch.isdigit() for ch in password):
        score += 1

    if any(not ch.isalnum() for ch in password):
        score += 1

    return score


def show_result(score):

    if score == 5:
        print("Very Strong Password")

    elif score >= 3:
        print("Strong Password")

    elif score == 2:
        print("Medium Password")

    else:
        print("Weak Password")


password = input("Enter Password: ")

result = check_password(password)

show_result(result)
```

```python
def fibonacci(limit):

    first = 0
    second = 1

    print("Fibonacci Series:")

    while first <= limit:
        print(first, end=" ")

        next_number = first + second
        first = second
        second = next_number


def main():

    limit = int(input("Enter Limit: "))

    fibonacci(limit)


main()
```

```python
def count_vowels(text):

    vowels = "aeiouAEIOU"
    total = 0

    for letter in text:
        if letter in vowels:
            total += 1

    return total


def count_words(text):

    words = text.split()
    return len(words)


def longest_word(text):

    words = text.split()
    longest = ""

    for word in words:
        if len(word) > len(longest):
            longest = word

    return longest


def main():

    sentence = input("Enter a sentence: ")

    print("Total Words :", count_words(sentence))
    print("Total Vowels:", count_vowels(sentence))
    print("Longest Word:", longest_word(sentence))


main()
```

```python
def even_numbers(numbers):

    even = []

    for num in numbers:
        if num % 2 == 0:
            even.append(num)

    return even


def odd_numbers(numbers):

    odd = []

    for num in numbers:
        if num % 2 != 0:
            odd.append(num)

    return odd


def average(numbers):
    return sum(numbers) / len(numbers)


def main():

    numbers = list(map(int, input("Enter numbers: ").split()))

    print("Even :", even_numbers(numbers))
    print("Odd  :", odd_numbers(numbers))
    print("Average :", average(numbers))


main()
```

```python
from functools import reduce


def process_numbers(*numbers, operation="sum"):
    """Perform different operations on a list of numbers."""

    if not numbers:
        return "No numbers provided."

    operations = {
        "sum": lambda x: sum(x),
        "average": lambda x: sum(x) / len(x),
        "maximum": lambda x: max(x),
        "minimum": lambda x: min(x),
        "product": lambda x: reduce(lambda a, b: a * b, x),
    }

    if operation not in operations:
        return "Invalid operation."

    return operations[operation](numbers)


def apply_discount(price, discount):
    """Calculate discounted price."""

    def validate():
        if price < 0 or discount < 0 or discount > 100:
            raise ValueError("Invalid price or discount.")

    try:
        validate()
        final_price = price - (price * discount / 100)
        return round(final_price, 2)
    except ValueError as error:
        return error


def student_report(name, **marks):
    """Generate a student's report."""

    total = sum(marks.values())
    average = total / len(marks)

    if average >= 80:
        grade = "A+"
    elif average >= 70:
        grade = "A"
    elif average >= 60:
        grade = "B"
    elif average >= 50:
        grade = "C"
    else:
        grade = "F"

    return {
        "Name": name,
        "Subjects": marks,
        "Total": total,
        "Average": round(average, 2),
        "Grade": grade,
    }


# ---------------- Main Program ----------------

print("=== Number Processing ===")
print("Sum:", process_numbers(5, 10, 15, 20))
print("Average:", process_numbers(5, 10, 15, 20, operation="average"))
print("Product:", process_numbers(5, 10, 15, operation="product"))

print("\n=== Discount Calculation ===")
print("Final Price:", apply_discount(2500, 15))

print("\n=== Student Report ===")
report = student_report(
    "Sumaiya",
    Python=95,
    Database=88,
    AI=91,
    Networking=84,
)

for key, value in report.items():
    print(f"{key}: {value}")
```

```python
def get_marks():
    marks = []

    while True:
        try:
            total_subjects = int(input("How many subjects? "))
            if total_subjects <= 0:
                print("Enter a valid number.")
                continue
            break
        except ValueError:
            print("Please enter a number.")

    for i in range(total_subjects):
        while True:
            try:
                mark = float(input(f"Enter mark for Subject {i + 1}: "))
                if 0 <= mark <= 100:
                    marks.append(mark)
                    break
                else:
                    print("Marks must be between 0 and 100.")
            except ValueError:
                print("Invalid input.")

    return marks


def calculate_result(marks):
    total = sum(marks)
    average = total / len(marks)

    if average >= 80:
        grade = "A+"
    elif average >= 70:
        grade = "A"
    elif average >= 60:
        grade = "A-"
    elif average >= 50:
        grade = "B"
    elif average >= 40:
        grade = "C"
    else:
        grade = "F"

    return {
        "Total": total,
        "Average": round(average, 2),
        "Highest": max(marks),
        "Lowest": min(marks),
        "Grade": grade
    }


def display_report(name, result):
    print("\n------ Student Report ------")
    print("Name    :", name)
    print("Total   :", result["Total"])
    print("Average :", result["Average"])
    print("Highest :", result["Highest"])
    print("Lowest  :", result["Lowest"])
    print("Grade   :", result["Grade"])


def main():
    print("=== Student Result System ===")

    student_name = input("Enter student name: ")

    marks = get_marks()

    result = calculate_result(marks)

    display_report(student_name, result)


if __name__ == "__main__":
    main()
```

```python
def create_account(accounts):
    account_number = input("Enter Account Number: ")

    if account_number in accounts:
        print("Account already exists.\n")
        return

    name = input("Enter Account Holder Name: ")
    opening_balance = float(input("Enter Opening Balance: "))

    accounts[account_number] = {
        "name": name,
        "balance": opening_balance
    }

    print("Account created successfully.\n")


def deposit(accounts):
    account_number = input("Enter Account Number: ")

    if account_number not in accounts:
        print("Account not found.\n")
        return

    amount = float(input("Enter Deposit Amount: "))

    if amount <= 0:
        print("Invalid amount.\n")
        return

    accounts[account_number]["balance"] += amount
    print("Deposit successful.\n")


def withdraw(accounts):
    account_number = input("Enter Account Number: ")

    if account_number not in accounts:
        print("Account not found.\n")
        return

    amount = float(input("Enter Withdrawal Amount: "))

    if amount > accounts[account_number]["balance"]:
        print("Insufficient balance.\n")
        return

    accounts[account_number]["balance"] -= amount
    print("Withdrawal successful.\n")


def check_balance(accounts):
    account_number = input("Enter Account Number: ")

    if account_number not in accounts:
        print("Account not found.\n")
        return

    print("\n----- Account Details -----")
    print("Name:", accounts[account_number]["name"])
    print("Balance:", accounts[account_number]["balance"])


def show_all_accounts(accounts):
    if not accounts:
        print("No accounts available.\n")
        return

    print("\n========== All Accounts ==========")

    for account_number, details in accounts.items():
        print(f"Account Number : {account_number}")
        print(f"Account Holder : {details['name']}")
        print(f"Balance        : {details['balance']}")
        print("-" * 35)


def menu():
    accounts = {}

    while True:
        print("\n====== Bank Account Management ======")
        print("1. Create Account")
        print("2. Deposit Money")
        print("3. Withdraw Money")
        print("4. Check Balance")
        print("5. Show All Accounts")
        print("6. Exit")

        choice = input("Enter your choice: ")

        if choice == "1":
            create_account(accounts)
        elif choice == "2":
            deposit(accounts)
        elif choice == "3":
            withdraw(accounts)
        elif choice == "4":
            check_balance(accounts)
        elif choice == "5":
            show_all_accounts(accounts)
        elif choice == "6":
            print("Thank you!")
            break
        else:
            print("Invalid choice. Try again.")


menu()
```

```python
def process_numbers(numbers, operation):
    results = []

    for number in numbers:
        result = operation(number)

        if result is not None:
            results.append(result)

    return results


def square_if_even(number):
    if number % 2 == 0:
        return number ** 2
    return None


def double_if_odd(number):
    if number % 2 != 0:
        return number * 2
    return None


def filter_and_transform(numbers, condition, transform):
    output = []

    for number in numbers:
        if condition(number):
            output.append(transform(number))

    return output


numbers = [3, 6, 9, 12, 15, 18, 21, 24]

even_squared = process_numbers(numbers, square_if_even)

odd_doubled = process_numbers(numbers, double_if_odd)

large_numbers = filter_and_transform(
    numbers,
    lambda x: x > 10,
    lambda x: x ** 2 + 5
)

print("Even numbers squared:", even_squared)
print("Odd numbers doubled:", odd_doubled)
print("Transformed values:", large_numbers)
```

```python
def fibonacci_analyzer(limit):
    cache = {}

    def fibonacci(n):
        if n in cache:
            return cache[n]

        if n <= 1:
            return n

        result = fibonacci(n - 1) + fibonacci(n - 2)
        cache[n] = result

        return result

    sequence = []

    for i in range(limit):
        value = fibonacci(i)

        sequence.append({
            "position": i,
            "value": value,
            "even": value % 2 == 0
        })

    return sequence


data = fibonacci_analyzer(12)

for item in data:
    status = "Even" if item["even"] else "Odd"

    print(
        f"Position: {item['position']} | "
        f"Value: {item['value']} | "
        f"{status}"
    )
```

```python
from functools import wraps
from time import perf_counter


def track_execution(func):
    @wraps(func)
    def wrapper(*args, **kwargs):
        start = perf_counter()

        try:
            result = func(*args, **kwargs)
            return result

        finally:
            end = perf_counter()
            elapsed = end - start

            print(
                f"\nFunction: {func.__name__}"
                f"\nExecution Time: {elapsed:.6f} seconds"
            )

    return wrapper


def validate_numbers(func):
    @wraps(func)
    def wrapper(*args, **kwargs):
        for value in args:
            if not isinstance(value, (int, float)):
                raise TypeError(
                    f"Invalid value: {value}"
                )

        return func(*args, **kwargs)

    return wrapper


@track_execution
@validate_numbers
def calculate_statistics(*numbers):
    if not numbers:
        return {"error": "No numbers provided"}

    total = sum(numbers)
    average = total / len(numbers)

    return {
        "count": len(numbers),
        "total": total,
        "average": round(average, 2),
        "maximum": max(numbers),
        "minimum": min(numbers),
        "range": max(numbers) - min(numbers)
    }


result = calculate_statistics(
    45, 78, 92, 63, 81, 56, 99
)

print("\nStatistics")
print("-" * 25)

for key, value in result.items():
    print(f"{key.capitalize():10}: {value}")
```

```python
def create_command_system():
    commands = {}

    def register(name):
        def decorator(func):
            commands[name] = func
            return func

        return decorator

    def run(command, *args, **kwargs):
        if command not in commands:
            return f"Unknown command: {command}"

        return commands[command](*args, **kwargs)

    def available_commands():
        return list(commands.keys())

    return register, run, available_commands


register, run, available = create_command_system()


@register("add")
def add_numbers(a, b):
    return a + b


@register("multiply")
def multiply_numbers(a, b):
    return a * b


@register("power")
def power_number(number, exponent=2):
    return number ** exponent


@register("average")
def calculate_average(*numbers):
    if not numbers:
        return 0

    return round(sum(numbers) / len(numbers), 2)


print("Available Commands:")
print(available())

print("\nResults:")
print("Add:", run("add", 25, 15))
print("Multiply:", run("multiply", 8, 7))
print("Power:", run("power", 5, 3))
print("Average:", run("average", 10, 20, 30, 40))
print("Test:", run("subtract", 10, 5))
```

```python
def task_manager(tasks):
    
    def get_priority(task):
        priority = task["priority"].lower()

        levels = {
            "high": 3,
            "medium": 2,
            "low": 1
        }

        return levels.get(priority, 0)

    def sort_tasks(items):
        return sorted(
            items,
            key=lambda task: (
                get_priority(task),
                task["deadline"]
            ),
            reverse=True
        )

    def calculate_progress(items):
        if not items:
            return 0

        completed = sum(
            task["completed"]
            for task in items
        )

        return round(
            (completed / len(items)) * 100,
            2
        )

    def create_summary(items):
        return {
            "total": len(items),
            "completed": sum(
                task["completed"] for task in items
            ),
            "pending": sum(
                not task["completed"] for task in items
            ),
            "progress": f"{calculate_progress(items)}%"
        }

    organized_tasks = sort_tasks(tasks)

    return {
        "tasks": organized_tasks,
        "summary": create_summary(organized_tasks)
    }


tasks = [
    {
        "title": "Complete ML Assignment",
        "priority": "High",
        "deadline": "2026-08-10",
        "completed": False
    },
    {
        "title": "Read Research Paper",
        "priority": "Medium",
        "deadline": "2026-08-12",
        "completed": True
    },
    {
        "title": "Practice Python",
        "priority": "Low",
        "deadline": "2026-08-15",
        "completed": False
    },
    {
        "title": "Prepare Presentation",
        "priority": "High",
        "deadline": "2026-08-09",
        "completed": True
    }
]


result = task_manager(tasks)

print("===== TASK LIST =====")

for task in result["tasks"]:
    status = "Done" if task["completed"] else "Pending"

    print(
        f"{task['title']} | "
        f"Priority: {task['priority']} | "
        f"Deadline: {task['deadline']} | "
        f"{status}"
    )

print("\n===== SUMMARY =====")

for key, value in result["summary"].items():
    print(f"{key.capitalize()}: {value}")
```

```python
def smart_number_generator(start, end, mode="prime"):
    cache = {}

    def is_prime(number):
        if number < 2:
            return False

        if number in cache:
            return cache[number]

        for i in range(2, int(number ** 0.5) + 1):
            if number % i == 0:
                cache[number] = False
                return False

        cache[number] = True
        return True

    def is_perfect_square(number):
        root = int(number ** 0.5)
        return root * root == number

    def generate():
        for number in range(start, end + 1):

            if mode == "prime":
                if is_prime(number):
                    yield number

            elif mode == "square":
                if is_perfect_square(number):
                    yield number

            elif mode == "both":
                if is_prime(number) or is_perfect_square(number):
                    yield number

            else:
                raise ValueError(
                    "Mode must be prime, square or both"
                )

    return generate


prime_numbers = smart_number_generator(
    1, 50, "prime"
)

square_numbers = smart_number_generator(
    1, 50, "square"
)

combined_numbers = smart_number_generator(
    1, 50, "both"
)


print("Prime Numbers:")
print(list(prime_numbers()))

print("\nPerfect Squares:")
print(list(square_numbers()))

print("\nPrime or Perfect Square:")
print(list(combined_numbers()))
```

```python
def transaction_analyzer(transactions, initial_balance=0):

    def calculate_balance(items):
        balance = initial_balance

        for item in items:
            if item["type"] == "deposit":
                balance += item["amount"]
            elif item["type"] == "withdraw":
                balance -= item["amount"]

        return balance

    def total_by_type(items, transaction_type):
        return sum(
            item["amount"]
            for item in items
            if item["type"] == transaction_type
        )

    def find_largest(items, transaction_type):
        filtered = [
            item for item in items
            if item["type"] == transaction_type
        ]

        if not filtered:
            return None

        return max(
            filtered,
            key=lambda item: item["amount"]
        )

    def create_summary(items):
        deposits = total_by_type(items, "deposit")
        withdrawals = total_by_type(items, "withdraw")

        return {
            "total_deposits": deposits,
            "total_withdrawals": withdrawals,
            "transaction_count": len(items),
            "final_balance": calculate_balance(items),
            "largest_deposit": find_largest(
                items, "deposit"
            ),
            "largest_withdrawal": find_largest(
                items, "withdraw"
            )
        }

    return create_summary(transactions)


transactions = [
    {"type": "deposit", "amount": 15000},
    {"type": "withdraw", "amount": 3000},
    {"type": "deposit", "amount": 8500},
    {"type": "withdraw", "amount": 2500},
    {"type": "deposit", "amount": 12000},
    {"type": "withdraw", "amount": 4500}
]


report = transaction_analyzer(
    transactions,
    initial_balance=5000
)


print("===== BANK REPORT =====")

for key, value in report.items():

    if isinstance(value, dict):
        print(
            f"{key}: "
            f"{value['amount']} "
            f"({value['type']})"
        )
    else:
        print(f"{key}: {value}")
```

```python
def analyze_structure(data, level=0):
    
    def get_indent(depth):
        return "    " * depth

    def process_item(name, content, depth):
        if isinstance(content, dict):
            print(f"{get_indent(depth)}📁 {name}/")

            for child_name, child_content in content.items():
                process_item(
                    child_name,
                    child_content,
                    depth + 1
                )
        else:
            print(
                f"{get_indent(depth)}📄 "
                f"{name} ({content} KB)"
            )

    total_size = 0

    def calculate_size(content):
        nonlocal total_size

        if isinstance(content, dict):
            for value in content.values():
                calculate_size(value)
        else:
            total_size += content

    calculate_size(data)

    print("===== FILE STRUCTURE =====\n")

    for name, content in data.items():
        process_item(name, content, level)

    return total_size


file_system = {
    "Project": {
        "main.py": 12,
        "config.py": 5,
        "src": {
            "model.py": 18,
            "utils.py": 9,
            "data": {
                "train.csv": 250,
                "test.csv": 80
            }
        },
        "README.md": 7
    }
}


total = analyze_structure(file_system)

print("\n==========================")
print(f"Total Size: {total} KB")
```

```python

def calculate_average(marks):
    total = sum(marks)
    return total / len(marks)


def calculate_grade(average):
    if average >= 80:
        return "A+"
    elif average >= 70:
        return "A"
    elif average >= 60:
        return "B"
    elif average >= 50:
        return "C"
    elif average >= 40:
        return "D"
    else:
        return "F"


def check_status(average):
    return "Passed" if average >= 40 else "Failed"


def create_student(name, student_id, department, marks):
    average = calculate_average(marks)

    return {
        "name": name,
        "id": student_id,
        "department": department,
        "marks": marks,
        "average": average,
        "grade": calculate_grade(average),
        "status": check_status(average)
    }


def display_student(student):
    print("\n-----------------------------")
    print("Student Information")
    print("-----------------------------")
    print("Name       :", student["name"])
    print("Student ID :", student["id"])
    print("Department :", student["department"])
    print("Marks      :", student["marks"])
    print("Average    :", round(student["average"], 2))
    print("Grade      :", student["grade"])
    print("Status     :", student["status"])


def find_top_student(students):
    if not students:
        return None

    return max(students, key=lambda student: student["average"])


def department_average(students, department):
    selected = [
        student for student in students
        if student["department"].lower() == department.lower()
    ]

    if not selected:
        return 0

    total = sum(student["average"] for student in selected)
    return total / len(selected)


def search_student(students, student_id):
    for student in students:
        if student["id"] == student_id:
            return student

    return None


students = []

students.append(
    create_student(
        "Sumaiya",
        "CSE101",
        "CSE",
        [85, 78, 92, 88, 90]
    )
)

students.append(
    create_student(
        "Rahim",
        "CSE102",
        "CSE",
        [72, 68, 75, 70, 80]
    )
)

students.append(
    create_student(
        "Nila",
        "CSE103",
        "CSE",
        [91, 89, 94, 87, 95]
    )
)

students.append(
    create_student(
        "Sadia",
        "EEE201",
        "EEE",
        [65, 72, 68, 70, 75]
    )
)


# Display all students
for student in students:
    display_student(student)


# Find the top student
top_student = find_top_student(students)

print("\n==============================")
print("TOP STUDENT")
print("==============================")

print("Name   :", top_student["name"])
print("Average:", round(top_student["average"], 2))
print("Grade  :", top_student["grade"])


# Calculate department average
cse_avg = department_average(students, "CSE")

print("\nCSE Department Average:",
      round(cse_avg, 2))


# Search for a particular student
student = search_student(students, "CSE102")

if student:
    print("\nStudent Found:")
    display_student(student)
else:
    print("\nStudent not found.")
```

```python
from functools import wraps
from datetime import datetime


# -----------------------------------------
# 1. Basic Expense Function
# -----------------------------------------

def add_expense(expenses, title, amount, category):
    expense = {
        "title": title,
        "amount": amount,
        "category": category,
        "date": datetime.now().strftime("%Y-%m-%d")
    }

    expenses.append(expense)
    return expense


# -----------------------------------------
# 2. Calculate Total
# -----------------------------------------

def calculate_total(expenses):
    return sum(expense["amount"] for expense in expenses)


# -----------------------------------------
# 3. Category Based Calculation
# -----------------------------------------

def category_total(expenses, category):
    return sum(
        expense["amount"]
        for expense in expenses
        if expense["category"].lower() == category.lower()
    )


# -----------------------------------------
# 4. Find Highest Expense
# -----------------------------------------

def highest_expense(expenses):

    if not expenses:
        return None

    return max(
        expenses,
        key=lambda expense: expense["amount"]
    )


# -----------------------------------------
# 5. Filter Expenses
# -----------------------------------------

def filter_expenses(expenses, minimum_amount=0):

    return [
        expense
        for expense in expenses
        if expense["amount"] >= minimum_amount
    ]


# -----------------------------------------
# 6. Expense Report
# -----------------------------------------

def generate_report(expenses):

    print("\n" + "=" * 45)
    print("             EXPENSE REPORT")
    print("=" * 45)

    for index, expense in enumerate(expenses, start=1):

        print(
            f"{index}. {expense['title']} | "
            f"{expense['category']} | "
            f"${expense['amount']:.2f} | "
            f"{expense['date']}"
        )

    print("-" * 45)
    print(f"Total Expense: ${calculate_total(expenses):.2f}")


# -----------------------------------------
# 7. Decorator
# -----------------------------------------

def log_function(func):

    @wraps(func)
    def wrapper(*args, **kwargs):

        print(f"\nRunning function: {func.__name__}")

        result = func(*args, **kwargs)

        print(f"Function completed: {func.__name__}")

        return result

    return wrapper


# -----------------------------------------
# 8. Decorated Function
# -----------------------------------------

@log_function
def show_category(expenses, category):

    total = category_total(expenses, category)

    print(
        f"{category.title()} expenses: "
        f"${total:.2f}"
    )


# -----------------------------------------
# 9. Closure Function
# -----------------------------------------

def create_budget_checker(budget):

    def check_budget(expenses):

        total = calculate_total(expenses)

        if total > budget:
            return (
                f"Budget exceeded by "
                f"${total - budget:.2f}"
            )

        remaining = budget - total

        return (
            f"Budget remaining: "
            f"${remaining:.2f}"
        )

    return check_budget


# -----------------------------------------
# 10. Recursive Function
# -----------------------------------------

def recursive_total(numbers):

    if not numbers:
        return 0

    return numbers[0] + recursive_total(numbers[1:])


# -----------------------------------------
# 11. Dynamic Summary Function
# -----------------------------------------

def summary(expenses, *categories, **options):

    print("\n========== SUMMARY ==========")

    for category in categories:

        total = category_total(expenses, category)

        print(
            f"{category.title():<15}: "
            f"${total:.2f}"
        )

    if options.get("show_highest"):

        highest = highest_expense(expenses)

        if highest:
            print(
                f"\nHighest Expense: "
                f"{highest['title']} "
                f"(${highest['amount']:.2f})"
            )

    if options.get("show_total"):

        print(
            f"Total: "
            f"${calculate_total(expenses):.2f}"
        )


# -----------------------------------------
# Main Program
# -----------------------------------------

expenses = []


# Adding expenses
add_expense(
    expenses,
    "Pizza",
    850,
    "Food"
)

add_expense(
    expenses,
    "Bus Ticket",
    120,
    "Transport"
)

add_expense(
    expenses,
    "Python Course",
    2500,
    "Education"
)

add_expense(
    expenses,
    "Coffee",
    300,
    "Food"
)

add_expense(
    expenses,
    "Movie",
    700,
    "Entertainment"
)

add_expense(
    expenses,
    "Burger",
    650,
    "Food"
)


# Display complete report
generate_report(expenses)


# Category calculation
show_category(
    expenses,
    "Food"
)


show_category(
    expenses,
    "Education"
)


# Find highest expense
highest = highest_expense(expenses)

print("\nHighest Expense:")
print(
    highest["title"],
    "->",
    highest["amount"]
)


# Filter expensive items
expensive_items = filter_expenses(
    expenses,
    minimum_amount=700
)

print("\nExpenses >= $700:")

for item in expensive_items:

    print(
        item["title"],
        "->",
        item["amount"]
    )


# Create a budget checker
budget_checker = create_budget_checker(5000)

print("\nBudget Status:")
print(
    budget_checker(expenses)
)


# Recursive calculation
amounts = [
    100,
    200,
    300,
    400,
    500
]

print(
    "\nRecursive Total:",
    recursive_total(amounts)
)


# Dynamic summary
summary(
    expenses,
    "Food",
    "Transport",
    "Education",
    "Entertainment",
    show_highest=True,
    show_total=True
)
```

```python
from datetime import datetime, timedelta
from functools import wraps


# ==========================================
# TASK DATA
# ==========================================

tasks = []


# ==========================================
# DECORATOR
# ==========================================

def log_task_action(func):

    @wraps(func)
    def wrapper(*args, **kwargs):

        print(f"\n>>> {func.__name__} started")

        result = func(*args, **kwargs)

        print(f">>> {func.__name__} completed")

        return result

    return wrapper


# ==========================================
# ADD TASK
# ==========================================

@log_task_action
def add_task(title, priority, days):

    deadline = datetime.now() + timedelta(days=days)

    task = {
        "title": title,
        "priority": priority,
        "deadline": deadline,
        "completed": False
    }

    tasks.append(task)

    print(
        f"Task '{title}' added successfully."
    )


# ==========================================
# COMPLETE TASK
# ==========================================

def complete_task(title):

    for task in tasks:

        if task["title"].lower() == title.lower():

            task["completed"] = True

            print(
                f"'{task['title']}' marked as completed."
            )

            return

    print("Task not found.")


# ==========================================
# PRIORITY SCORE
# ==========================================

def priority_score(priority):

    scores = {
        "low": 1,
        "medium": 2,
        "high": 3,
        "urgent": 4
    }

    return scores.get(
        priority.lower(),
        0
    )


# ==========================================
# SORT TASKS
# ==========================================

def sort_tasks():

    return sorted(
        tasks,
        key=lambda task:
        priority_score(task["priority"]),
        reverse=True
    )


# ==========================================
# FIND OVERDUE TASKS
# ==========================================

def overdue_tasks():

    now = datetime.now()

    return [
        task
        for task in tasks
        if task["deadline"] < now
        and not task["completed"]
    ]


# ==========================================
# TASK STATISTICS
# ==========================================

def task_statistics():

    total = len(tasks)

    completed = sum(
        task["completed"]
        for task in tasks
    )

    pending = total - completed

    overdue = len(
        overdue_tasks()
    )

    print("\n========== TASK STATISTICS ==========")

    print("Total Tasks     :", total)
    print("Completed Tasks :", completed)
    print("Pending Tasks   :", pending)
    print("Overdue Tasks   :", overdue)

    if total > 0:

        completion_rate = (
            completed / total
        ) * 100

        print(
            "Completion Rate :",
            f"{completion_rate:.2f}%"
        )


# ==========================================
# SEARCH TASK
# ==========================================

def search_task(keyword):

    keyword = keyword.lower()

    results = [
        task
        for task in tasks
        if keyword in task["title"].lower()
    ]

    return results


# ==========================================
# CREATE PRIORITY FILTER
# ==========================================

def create_priority_filter(priority):

    def filter_tasks():

        return [
            task
            for task in tasks
            if task["priority"].lower()
            == priority.lower()
        ]

    return filter_tasks


# ==========================================
# DISPLAY TASKS
# ==========================================

def display_tasks(task_list):

    print("\n========== TASK LIST ==========")

    if not task_list:
        print("No tasks found.")
        return

    for number, task in enumerate(
        task_list,
        start=1
    ):

        status = (
            "Completed"
            if task["completed"]
            else "Pending"
        )

        deadline = task["deadline"].strftime(
            "%Y-%m-%d"
        )

        print(
            f"{number}. "
            f"{task['title']} | "
            f"Priority: {task['priority']} | "
            f"Deadline: {deadline} | "
            f"Status: {status}"
        )


# ==========================================
# PRODUCTIVITY SCORE
# ==========================================

def productivity_score():

    if not tasks:
        return 0

    completed = sum(
        task["completed"]
        for task in tasks
    )

    urgent_tasks = sum(
        task["priority"].lower()
        == "urgent"
        for task in tasks
    )

    score = (
        completed * 20
        + urgent_tasks * 5
    )

    return min(score, 100)


# ==========================================
# ADD TASKS
# ==========================================

add_task(
    "Complete Python Assignment",
    "high",
    3
)

add_task(
    "Read Research Paper",
    "medium",
    5
)

add_task(
    "Prepare Presentation",
    "urgent",
    2
)

add_task(
    "Practice Machine Learning",
    "high",
    4
)

add_task(
    "Organize GitHub Repository",
    "low",
    7
)


# ==========================================
# DISPLAY ALL TASKS
# ==========================================

display_tasks(tasks)


# ==========================================
# COMPLETE SOME TASKS
# ==========================================

complete_task(
    "Complete Python Assignment"
)

complete_task(
    "Read Research Paper"
)


# ==========================================
# SORT BY PRIORITY
# ==========================================

sorted_task_list = sort_tasks()

print(
    "\n========== PRIORITY ORDER =========="
)

display_tasks(sorted_task_list)


# ==========================================
# SEARCH
# ==========================================

print(
    "\n========== SEARCH RESULT =========="
)

results = search_task("Python")

display_tasks(results)


# ==========================================
# PRIORITY FILTER
# ==========================================

high_priority = create_priority_filter(
    "High"
)

high_tasks = high_priority()

print(
    "\n========== HIGH PRIORITY TASKS =========="
)

display_tasks(high_tasks)


# ==========================================
# OVERDUE TASKS
# ==========================================

print(
    "\n========== OVERDUE TASKS =========="
)

display_tasks(
    overdue_tasks()
)


# ==========================================
# STATISTICS
# ==========================================

task_statistics()


# ==========================================
# PRODUCTIVITY SCORE
# ==========================================

print(
    "\nProductivity Score:",
    productivity_score(),
    "/ 100"
)
```

```python
from functools import reduce
from collections import Counter


# ==========================================
# MOVIE DATA
# ==========================================

movies = [
    {
        "title": "Inception",
        "genre": ["Sci-Fi", "Thriller"],
        "rating": 8.8,
        "year": 2010,
        "language": "English"
    },
    {
        "title": "Interstellar",
        "genre": ["Sci-Fi", "Drama"],
        "rating": 8.7,
        "year": 2014,
        "language": "English"
    },
    {
        "title": "The Dark Knight",
        "genre": ["Action", "Crime"],
        "rating": 9.0,
        "year": 2008,
        "language": "English"
    },
    {
        "title": "Parasite",
        "genre": ["Drama", "Thriller"],
        "rating": 8.5,
        "year": 2019,
        "language": "Korean"
    },
    {
        "title": "3 Idiots",
        "genre": ["Comedy", "Drama"],
        "rating": 8.4,
        "year": 2009,
        "language": "Hindi"
    },
    {
        "title": "Avengers: Endgame",
        "genre": ["Action", "Sci-Fi"],
        "rating": 8.4,
        "year": 2019,
        "language": "English"
    },
    {
        "title": "The Matrix",
        "genre": ["Sci-Fi", "Action"],
        "rating": 8.7,
        "year": 1999,
        "language": "English"
    }
]


# ==========================================
# BASIC FILTER FUNCTION
# ==========================================

def filter_by_rating(movie_list, minimum_rating):

    return [
        movie
        for movie in movie_list
        if movie["rating"] >= minimum_rating
    ]


# ==========================================
# GENRE SEARCH
# ==========================================

def find_by_genre(movie_list, genre):

    genre = genre.lower()

    return [
        movie
        for movie in movie_list
        if any(
            genre == g.lower()
            for g in movie["genre"]
        )
    ]


# ==========================================
# LANGUAGE FILTER
# ==========================================

def find_by_language(movie_list, language):

    return [
        movie
        for movie in movie_list
        if movie["language"].lower()
        == language.lower()
    ]


# ==========================================
# SORT MOVIES
# ==========================================

def sort_movies(movie_list, key_name):

    return sorted(
        movie_list,
        key=lambda movie: movie[key_name],
        reverse=True
    )


# ==========================================
# AVERAGE RATING
# ==========================================

def average_rating(movie_list):

    if not movie_list:
        return 0

    total = sum(
        movie["rating"]
        for movie in movie_list
    )

    return total / len(movie_list)


# ==========================================
# MOST COMMON GENRE
# ==========================================

def most_popular_genre(movie_list):

    all_genres = []

    for movie in movie_list:
        all_genres.extend(movie["genre"])

    if not all_genres:
        return None

    counter = Counter(all_genres)

    return counter.most_common(1)[0]


# ==========================================
# RECOMMENDATION SCORE
# ==========================================

def recommendation_score(movie, preferred_genres):

    score = movie["rating"]

    for genre in movie["genre"]:

        if genre.lower() in [
            g.lower()
            for g in preferred_genres
        ]:
            score += 1.5

    if movie["year"] >= 2015:
        score += 0.5

    return score


# ==========================================
# SMART RECOMMENDER
# ==========================================

def recommend_movies(
    movie_list,
    preferred_genres,
    minimum_rating=7.5
):

    candidates = filter_by_rating(
        movie_list,
        minimum_rating
    )

    ranked_movies = sorted(
        candidates,
        key=lambda movie:
        recommendation_score(
            movie,
            preferred_genres
        ),
        reverse=True
    )

    return ranked_movies


# ==========================================
# DISPLAY FUNCTION
# ==========================================

def display_movies(movie_list):

    if not movie_list:
        print("No movies found.")
        return

    print("\n" + "=" * 60)

    for index, movie in enumerate(
        movie_list,
        start=1
    ):

        genres = ", ".join(
            movie["genre"]
        )

        print(
            f"{index}. {movie['title']}"
        )

        print(
            f"   Rating   : {movie['rating']}"
        )

        print(
            f"   Genre    : {genres}"
        )

        print(
            f"   Year     : {movie['year']}"
        )

        print(
            f"   Language : {movie['language']}"
        )

    print("=" * 60)


# ==========================================
# FUNCTION COMPOSITION
# ==========================================

def create_movie_filter(
    minimum_rating,
    language=None
):

    def apply_filter(movie_list):

        result = filter_by_rating(
            movie_list,
            minimum_rating
        )

        if language:
            result = find_by_language(
                result,
                language
            )

        return result

    return apply_filter


# ==========================================
# MOVIE STATISTICS
# ==========================================

def movie_statistics(movie_list):

    print("\n========== MOVIE STATISTICS ==========")

    print(
        "Total Movies:",
        len(movie_list)
    )

    print(
        "Average Rating:",
        round(
            average_rating(movie_list),
            2
        )
    )

    popular = most_popular_genre(
        movie_list
    )

    if popular:
        print(
            "Popular Genre:",
            popular[0]
        )

        print(
            "Genre Count:",
            popular[1]
        )


# ==========================================
# MAIN PROGRAM
# ==========================================

print("\nALL MOVIES")
display_movies(movies)


# High-rated movies
high_rated = filter_by_rating(
    movies,
    8.5
)

print("\nHIGH RATED MOVIES")
display_movies(high_rated)


# Sci-Fi movies
sci_fi_movies = find_by_genre(
    movies,
    "Sci-Fi"
)

print("\nSCI-FI MOVIES")
display_movies(sci_fi_movies)


# Sort by rating
sorted_movies = sort_movies(
    movies,
    "rating"
)

print("\nMOVIES SORTED BY RATING")
display_movies(sorted_movies)


# Smart recommendations
preferences = [
    "Sci-Fi",
    "Action",
    "Thriller"
]

recommendations = recommend_movies(
    movies,
    preferences,
    minimum_rating=8.0
)

print("\nSMART RECOMMENDATIONS")
display_movies(recommendations)


# Closure-based filter
english_filter = create_movie_filter(
    minimum_rating=8.5,
    language="English"
)

english_movies = english_filter(
    movies
)

print(
    "\nHIGH-RATED ENGLISH MOVIES"
)

display_movies(
    english_movies
)


# Statistics
movie_statistics(movies)


# Final recommendation
if recommendations:

    best_movie = recommendations[0]

    print(
        "\nRecommended Movie:",
        best_movie["title"]
    )

    print(
        "Recommendation Score:",
        round(
            recommendation_score(
                best_movie,
                preferences
            ),
            2
        )
    )
```

```python
from datetime import datetime


def add_expense(expenses, name, amount, category):
    expense = {
        "name": name,
        "amount": amount,
        "category": category,
        "date": datetime.now().strftime("%d-%m-%Y")
    }

    expenses.append(expense)
    print(f"{name} added successfully.")


def show_expenses(expenses):
    print("\n------ My Expenses ------")

    if len(expenses) == 0:
        print("No expenses found.")
        return

    for i, expense in enumerate(expenses, 1):
        print(
            f"{i}. {expense['name']} - "
            f"{expense['amount']} BDT - "
            f"{expense['category']} - "
            f"{expense['date']}"
        )


def get_total(expenses):
    total = 0

    for expense in expenses:
        total += expense["amount"]

    return total


def category_expense(expenses, category):
    total = 0

    for expense in expenses:
        if expense["category"].lower() == category.lower():
            total += expense["amount"]

    return total


def find_expensive_item(expenses):
    if not expenses:
        return None

    expensive = expenses[0]

    for expense in expenses:
        if expense["amount"] > expensive["amount"]:
            expensive = expense

    return expensive


def search_expense(expenses, keyword):
    found = []

    for expense in expenses:
        if keyword.lower() in expense["name"].lower():
            found.append(expense)

    return found


def monthly_budget(expenses, budget):
    total = get_total(expenses)
    remaining = budget - total

    print("\n------ Budget Status ------")
    print("Budget:", budget, "BDT")
    print("Spent :", total, "BDT")

    if remaining > 0:
        print("Remaining:", remaining, "BDT")
    elif remaining == 0:
        print("You used your full budget.")
    else:
        print(
            "Budget exceeded by:",
            abs(remaining),
            "BDT"
        )


def expense_summary(expenses):
    print("\n------ Summary ------")

    total = get_total(expenses)

    print("Total expenses:", total, "BDT")

    food = category_expense(
        expenses,
        "Food"
    )

    transport = category_expense(
        expenses,
        "Transport"
    )

    shopping = category_expense(
        expenses,
        "Shopping"
    )

    print("Food:", food, "BDT")
    print("Transport:", transport, "BDT")
    print("Shopping:", shopping, "BDT")


def main():
    expenses = []

    add_expense(
        expenses,
        "Lunch",
        250,
        "Food"
    )

    add_expense(
        expenses,
        "Uber",
        180,
        "Transport"
    )

    add_expense(
        expenses,
        "T-shirt",
        850,
        "Shopping"
    )

    add_expense(
        expenses,
        "Coffee",
        150,
        "Food"
    )

    add_expense(
        expenses,
        "Bus",
        60,
        "Transport"
    )

    show_expenses(expenses)

    print(
        "\nTotal spent:",
        get_total(expenses),
        "BDT"
    )

    expensive = find_expensive_item(
        expenses
    )

    if expensive:
        print(
            "\nMost expensive:",
            expensive["name"],
            "-",
            expensive["amount"],
            "BDT"
        )

    results = search_expense(
        expenses,
        "coffee"
    )

    print("\nSearch result:")

    for item in results:
        print(
            item["name"],
            "->",
            item["amount"],
            "BDT"
        )

    monthly_budget(
        expenses,
        3000
    )

    expense_summary(expenses)


main()
```

```python
def add_food(food_list, name, calories, meal):
    food = {
        "name": name,
        "calories": calories,
        "meal": meal
    }

    food_list.append(food)
    print(name, "added successfully.")


def show_foods(food_list):
    print("\n----- Today's Food -----")

    if not food_list:
        print("No food has been added yet.")
        return

    for food in food_list:
        print(
            f"{food['meal']}: "
            f"{food['name']} - "
            f"{food['calories']} calories"
        )


def total_calories(food_list):
    total = 0

    for food in food_list:
        total += food["calories"]

    return total


def meal_calories(food_list, meal):
    total = 0

    for food in food_list:
        if food["meal"].lower() == meal.lower():
            total += food["calories"]

    return total


def find_highest_calorie_food(food_list):
    if not food_list:
        return None

    highest = food_list[0]

    for food in food_list:
        if food["calories"] > highest["calories"]:
            highest = food

    return highest


def search_food(food_list, keyword):
    result = []

    for food in food_list:
        if keyword.lower() in food["name"].lower():
            result.append(food)

    return result


def calorie_status(food_list, daily_limit):
    total = total_calories(food_list)

    print("\n----- Calorie Status -----")
    print("Daily limit:", daily_limit)
    print("Consumed:", total)

    if total < daily_limit:
        remaining = daily_limit - total
        print("Remaining:", remaining, "calories")

    elif total == daily_limit:
        print("You reached your daily limit.")

    else:
        extra = total - daily_limit
        print("Exceeded by:", extra, "calories")


def daily_summary(food_list):
    total = total_calories(food_list)

    breakfast = meal_calories(
        food_list,
        "Breakfast"
    )

    lunch = meal_calories(
        food_list,
        "Lunch"
    )

    dinner = meal_calories(
        food_list,
        "Dinner"
    )

    snacks = meal_calories(
        food_list,
        "Snack"
    )

    print("\n========== DAILY SUMMARY ==========")

    print("Breakfast:", breakfast)
    print("Lunch:", lunch)
    print("Dinner:", dinner)
    print("Snacks:", snacks)
    print("Total:", total)


def compare_days(today, yesterday):
    today_total = total_calories(today)
    yesterday_total = total_calories(yesterday)

    print("\n----- Comparison -----")

    print("Yesterday:", yesterday_total)
    print("Today:", today_total)

    difference = today_total - yesterday_total

    if difference > 0:
        print(
            "Today you consumed",
            difference,
            "more calories."
        )

    elif difference < 0:
        print(
            "Today you consumed",
            abs(difference),
            "fewer calories."
        )

    else:
        print(
            "Both days have the same calories."
        )


def main():

    today = []

    yesterday = [
        {
            "name": "Rice",
            "calories": 300,
            "meal": "Lunch"
        },
        {
            "name": "Chicken",
            "calories": 250,
            "meal": "Lunch"
        },
        {
            "name": "Apple",
            "calories": 95,
            "meal": "Snack"
        }
    ]

    add_food(
        today,
        "Egg",
        140,
        "Breakfast"
    )

    add_food(
        today,
        "Bread",
        160,
        "Breakfast"
    )

    add_food(
        today,
        "Chicken Rice",
        450,
        "Lunch"
    )

    add_food(
        today,
        "Coffee",
        80,
        "Snack"
    )

    add_food(
        today,
        "Pizza",
        350,
        "Dinner"
    )

    show_foods(today)

    print(
        "\nTotal calories:",
        total_calories(today)
    )

    highest = find_highest_calorie_food(today)

    if highest:
        print(
            "\nHighest calorie food:",
            highest["name"]
        )

        print(
            "Calories:",
            highest["calories"]
        )

    print("\nSearch result for 'chicken':")

    result = search_food(
        today,
        "chicken"
    )

    if result:
        for food in result:
            print(
                food["name"],
                "-",
                food["calories"]
            )
    else:
        print("No food found.")

    calorie_status(
        today,
        1800
    )

    daily_summary(today)

    compare_days(
        today,
        yesterday
    )


main()
```

```python
def add_product(cart, name, price, quantity):
    product = {
        "name": name,
        "price": price,
        "quantity": quantity
    }

    cart.append(product)
    print(name, "added to cart.")


def remove_product(cart, name):
    for product in cart:
        if product["name"].lower() == name.lower():
            cart.remove(product)
            print(name, "removed from cart.")
            return

    print("Product not found.")


def calculate_subtotal(cart):
    subtotal = 0

    for product in cart:
        subtotal += product["price"] * product["quantity"]

    return subtotal


def calculate_discount(subtotal):
    if subtotal >= 5000:
        return subtotal * 0.20
    elif subtotal >= 3000:
        return subtotal * 0.10
    elif subtotal >= 1500:
        return subtotal * 0.05
    else:
        return 0


def calculate_delivery_charge(subtotal):
    if subtotal >= 3000:
        return 0

    return 100


def calculate_final_bill(cart):
    subtotal = calculate_subtotal(cart)
    discount = calculate_discount(subtotal)
    delivery = calculate_delivery_charge(subtotal)

    final_amount = subtotal - discount + delivery

    return subtotal, discount, delivery, final_amount


def show_cart(cart):

    if not cart:
        print("Your cart is empty.")
        return

    print("\n========== YOUR CART ==========")

    for product in cart:
        total = product["price"] * product["quantity"]

        print(
            f"{product['name']} - "
            f"{product['quantity']} x "
            f"{product['price']} = {total}"
        )


def show_bill(cart):

    subtotal, discount, delivery, final_amount = \
        calculate_final_bill(cart)

    print("\n========== BILL ==========")

    print("Subtotal :", subtotal)
    print("Discount :", discount)
    print("Delivery :", delivery)
    print("--------------------------")
    print("Total    :", final_amount)


cart = []

add_product(cart, "Laptop Bag", 1800, 1)
add_product(cart, "Keyboard", 1200, 2)
add_product(cart, "Mouse", 700, 1)

show_cart(cart)

remove_product(cart, "Mouse")

show_cart(cart)

show_bill(cart)
```

```python
def add_doctor(doctors, doctor_id, name, specialization):
    doctor = {
        "id": doctor_id,
        "name": name,
        "specialization": specialization,
        "appointments": []
    }

    doctors.append(doctor)


def find_doctor(doctors, doctor_id):
    for doctor in doctors:
        if doctor["id"] == doctor_id:
            return doctor

    return None


def book_appointment(
    doctors,
    patients,
    doctor_id,
    patient_name,
    date,
    time
):
    doctor = find_doctor(doctors, doctor_id)

    if doctor is None:
        print("Doctor not found.")
        return False

    for appointment in doctor["appointments"]:
        if (
            appointment["date"] == date
            and appointment["time"] == time
        ):
            print("This time slot is already booked.")
            return False

    appointment = {
        "patient": patient_name,
        "date": date,
        "time": time
    }

    doctor["appointments"].append(appointment)

    patients.append({
        "name": patient_name,
        "doctor": doctor["name"],
        "date": date,
        "time": time
    })

    print("Appointment booked successfully.")
    return True


def cancel_appointment(
    doctors,
    doctor_id,
    patient_name,
    date
):
    doctor = find_doctor(doctors, doctor_id)

    if doctor is None:
        print("Doctor not found.")
        return False

    for appointment in doctor["appointments"]:

        if (
            appointment["patient"].lower()
            == patient_name.lower()
            and appointment["date"] == date
        ):
            doctor["appointments"].remove(appointment)

            print("Appointment cancelled.")
            return True

    print("Appointment not found.")
    return False


def show_doctor_appointments(doctor):
    print("\n==============================")
    print("Doctor:", doctor["name"])
    print("Specialization:", doctor["specialization"])
    print("==============================")

    if not doctor["appointments"]:
        print("No appointments.")
        return

    for appointment in doctor["appointments"]:
        print(
            "Patient:",
            appointment["patient"]
        )

        print(
            "Date:",
            appointment["date"]
        )

        print(
            "Time:",
            appointment["time"]
        )

        print("------------------------------")


def show_all_doctors(doctors):
    for doctor in doctors:
        print(
            doctor["id"],
            "-",
            doctor["name"],
            "-",
            doctor["specialization"]
        )


def search_doctor(doctors, specialization):
    result = []

    for doctor in doctors:

        if (
            doctor["specialization"].lower()
            == specialization.lower()
        ):
            result.append(doctor)

    return result


doctors = []
patients = []

add_doctor(
    doctors,
    101,
    "Dr. Rahman",
    "Cardiology"
)

add_doctor(
    doctors,
    102,
    "Dr. Nadia",
    "Dermatology"
)

add_doctor(
    doctors,
    103,
    "Dr. Karim",
    "Neurology"
)


print("Available Doctors")
show_all_doctors(doctors)


book_appointment(
    doctors,
    patients,
    101,
    "Sumaiya",
    "20-08-2026",
    "10:00 AM"
)

book_appointment(
    doctors,
    patients,
    101,
    "Rahim",
    "20-08-2026",
    "11:00 AM"
)

book_appointment(
    doctors,
    patients,
    102,
    "Karim",
    "21-08-2026",
    "12:00 PM"
)


doctor = find_doctor(doctors, 101)

if doctor:
    show_doctor_appointments(doctor)


print("\nDermatology Doctors:")

dermatologists = search_doctor(
    doctors,
    "Dermatology"
)

for doctor in dermatologists:
    print(
        doctor["name"],
        "-",
        doctor["specialization"]
    )


cancel_appointment(
    doctors,
    101,
    "Sumaiya",
    "20-08-2026"
)
```

```python
patients = []


def add_patient(patient_id, name, age, disease, doctor):
    patient = {
        "id": patient_id,
        "name": name,
        "age": age,
        "disease": disease,
        "doctor": doctor,
        "bill": 0,
        "status": "Admitted"
    }

    patients.append(patient)
    print(f"{name} has been admitted successfully.")


def find_patient(patient_id):
    for patient in patients:
        if patient["id"] == patient_id:
            return patient

    return None


def add_bill(patient_id, amount):
    patient = find_patient(patient_id)

    if patient is None:
        print("Patient not found.")
        return

    if amount <= 0:
        print("Invalid amount.")
        return

    patient["bill"] += amount

    print(
        f"Bill added successfully. "
        f"Current bill: {patient['bill']}"
    )


def discharge_patient(patient_id):
    patient = find_patient(patient_id)

    if patient is None:
        print("Patient not found.")
        return

    patient["status"] = "Discharged"

    print(
        f"{patient['name']} has been discharged."
    )


def show_patient(patient):
    print("\n-----------------------------")
    print("Patient ID :", patient["id"])
    print("Name       :", patient["name"])
    print("Age        :", patient["age"])
    print("Disease    :", patient["disease"])
    print("Doctor     :", patient["doctor"])
    print("Bill       :", patient["bill"])
    print("Status     :", patient["status"])
    print("-----------------------------")


def show_all_patients():
    if not patients:
        print("No patients available.")
        return

    for patient in patients:
        show_patient(patient)


def search_by_doctor(doctor_name):
    found = []

    for patient in patients:
        if patient["doctor"].lower() == doctor_name.lower():
            found.append(patient)

    return found


def calculate_total_bill():
    total = 0

    for patient in patients:
        total += patient["bill"]

    return total


def main():

    add_patient(
        101,
        "Rahim",
        45,
        "Fever",
        "Dr. Ahmed"
    )

    add_patient(
        102,
        "Karim",
        32,
        "Diabetes",
        "Dr. Rahman"
    )

    add_patient(
        103,
        "Nadia",
        28,
        "Fracture",
        "Dr. Ahmed"
    )

    add_bill(101, 1500)
    add_bill(101, 2500)

    add_bill(102, 3200)

    add_bill(103, 5000)

    discharge_patient(101)

    print("\nALL PATIENTS")
    show_all_patients()

    print("\nPatients of Dr. Ahmed:")

    result = search_by_doctor("Dr. Ahmed")

    for patient in result:
        show_patient(patient)

    print(
        "\nTotal Hospital Bill:",
        calculate_total_bill()
    )


if __name__ == "__main__":
    main()
```

```python
accounts = {
    1001: {
        "name": "Rahim",
        "pin": "1234",
        "balance": 25000,
        "transactions": []
    },

    1002: {
        "name": "Karim",
        "pin": "5678",
        "balance": 18000,
        "transactions": []
    }
}


def login(account_number, pin):
    if account_number not in accounts:
        return False

    if accounts[account_number]["pin"] != pin:
        return False

    return True


def check_balance(account_number):
    return accounts[account_number]["balance"]


def deposit(account_number, amount):

    if amount <= 0:
        print("Invalid deposit amount.")
        return

    account = accounts[account_number]

    account["balance"] += amount

    account["transactions"].append(
        f"Deposited {amount}"
    )

    print(
        f"{amount} deposited successfully."
    )


def withdraw(account_number, amount):

    account = accounts[account_number]

    if amount <= 0:
        print("Invalid amount.")
        return

    if amount > account["balance"]:
        print("Insufficient balance.")
        return

    account["balance"] -= amount

    account["transactions"].append(
        f"Withdrawn {amount}"
    )

    print(
        f"{amount} withdrawn successfully."
    )


def transfer(sender, receiver, amount):

    if receiver not in accounts:
        print("Receiver account not found.")
        return

    if amount <= 0:
        print("Invalid amount.")
        return

    if amount > accounts[sender]["balance"]:
        print("Insufficient balance.")
        return

    accounts[sender]["balance"] -= amount
    accounts[receiver]["balance"] += amount

    accounts[sender]["transactions"].append(
        f"Transferred {amount} to {receiver}"
    )

    accounts[receiver]["transactions"].append(
        f"Received {amount} from {sender}"
    )

    print("Transfer completed successfully.")


def show_transactions(account_number):

    transactions = accounts[
        account_number
    ]["transactions"]

    if not transactions:
        print("No transactions yet.")
        return

    print("\nTransaction History:")

    for transaction in transactions:
        print("-", transaction)


def atm_menu(account_number):

    while True:

        print("\n========== ATM ==========")
        print("1. Check Balance")
        print("2. Deposit")
        print("3. Withdraw")
        print("4. Transfer")
        print("5. Transaction History")
        print("6. Exit")

        choice = input("Choose an option: ")

        if choice == "1":

            balance = check_balance(
                account_number
            )

            print("Balance:", balance)

        elif choice == "2":

            amount = float(
                input("Enter deposit amount: ")
            )

            deposit(
                account_number,
                amount
            )

        elif choice == "3":

            amount = float(
                input("Enter withdrawal amount: ")
            )

            withdraw(
                account_number,
                amount
            )

        elif choice == "4":

            receiver = int(
                input("Receiver account: ")
            )

            amount = float(
                input("Enter amount: ")
            )

            transfer(
                account_number,
                receiver,
                amount
            )

        elif choice == "5":

            show_transactions(
                account_number
            )

        elif choice == "6":

            print("Thank you for using the ATM.")
            break

        else:
            print("Invalid choice.")


def main():

    print("========== BANK ATM ==========")

    account_number = int(
        input("Account Number: ")
    )

    pin = input("PIN: ")

    if login(account_number, pin):

        print(
            f"\nWelcome, "
            f"{accounts[account_number]['name']}!"
        )

        atm_menu(account_number)

    else:
        print("Invalid account number or PIN.")


if __name__ == "__main__":
    main()
```

```python
def create_room(room_number, room_type, price):
    return {
        "number": room_number,
        "type": room_type,
        "price": price,
        "available": True,
        "guest": None
    }


def find_room(rooms, room_number):
    for room in rooms:
        if room["number"] == room_number:
            return room

    return None


def show_available_rooms(rooms):
    print("\nAvailable Rooms")
    print("------------------------")

    for room in rooms:
        if room["available"]:
            print(
                f"Room {room['number']} | "
                f"{room['type']} | "
                f"Tk {room['price']} per night"
            )


def book_room(
    rooms,
    room_number,
    guest_name,
    nights
):
    room = find_room(rooms, room_number)

    if room is None:
        print("Room does not exist.")
        return None

    if not room["available"]:
        print("Room is already booked.")
        return None

    if nights <= 0:
        print("Invalid number of nights.")
        return None

    total = room["price"] * nights

    room["available"] = False

    room["guest"] = {
        "name": guest_name,
        "nights": nights,
        "total": total
    }

    print(
        f"\nRoom {room_number} booked successfully."
    )

    print("Guest:", guest_name)
    print("Nights:", nights)
    print("Total:", total)

    return total


def cancel_booking(rooms, room_number):

    room = find_room(rooms, room_number)

    if room is None:
        print("Room not found.")
        return False

    if room["available"]:
        print("This room is not booked.")
        return False

    room["available"] = True
    room["guest"] = None

    print(
        f"Booking for room {room_number} cancelled."
    )

    return True


def hotel_revenue(rooms):

    total = 0

    for room in rooms:

        if not room["available"]:
            total += room["guest"]["total"]

    return total


def show_booked_rooms(rooms):

    print("\nBooked Rooms")
    print("------------------------")

    for room in rooms:

        if not room["available"]:

            print(
                f"Room: {room['number']}"
            )

            print(
                f"Guest: {room['guest']['name']}"
            )

            print(
                f"Nights: {room['guest']['nights']}"
            )

            print(
                f"Bill: Tk {room['guest']['total']}"
            )

            print("------------------------")


rooms = [
    create_room(101, "Single", 1500),
    create_room(102, "Double", 2500),
    create_room(103, "Deluxe", 4000),
    create_room(104, "Suite", 6000),
    create_room(105, "Single", 1500)
]


show_available_rooms(rooms)

book_room(
    rooms,
    103,
    "Sumaiya",
    3
)

book_room(
    rooms,
    102,
    "Rahim",
    2
)

book_room(
    rooms,
    104,
    "Nadia",
    4
)

show_booked_rooms(rooms)

print(
    "\nCurrent Hotel Revenue:",
    hotel_revenue(rooms)
)

cancel_booking(rooms, 102)

print(
    "\nRevenue after cancellation:",
    hotel_revenue(rooms)
)
```

```python
def add_book(library, book_id, title, author):
    book = {
        "id": book_id,
        "title": title,
        "author": author,
        "available": True,
        "borrowed_by": None
    }

    library.append(book)


def search_by_title(library, keyword):

    results = []

    for book in library:

        if keyword.lower() in book["title"].lower():
            results.append(book)

    return results


def search_by_author(library, author):

    results = []

    for book in library:

        if author.lower() in book["author"].lower():
            results.append(book)

    return results


def find_book(library, book_id):

    for book in library:

        if book["id"] == book_id:
            return book

    return None


def borrow_book(library, book_id, student):

    book = find_book(library, book_id)

    if book is None:
        print("Book not found.")
        return False

    if not book["available"]:
        print(
            f"Book already borrowed by "
            f"{book['borrowed_by']}"
        )

        return False

    book["available"] = False
    book["borrowed_by"] = student

    print(
        f"{student} borrowed "
        f"'{book['title']}'."
    )

    return True


def return_book(library, book_id):

    book = find_book(library, book_id)

    if book is None:
        print("Book not found.")
        return False

    if book["available"]:
        print("This book is already available.")
        return False

    print(
        f"'{book['title']}' returned by "
        f"{book['borrowed_by']}."
    )

    book["available"] = True
    book["borrowed_by"] = None

    return True


def show_library(library):

    print("\n========== LIBRARY ==========")

    for book in library:

        status = (
            "Available"
            if book["available"]
            else f"Borrowed by {book['borrowed_by']}"
        )

        print(
            f"{book['id']} | "
            f"{book['title']} | "
            f"{book['author']} | "
            f"{status}"
        )


def count_available_books(library):

    count = 0

    for book in library:

        if book["available"]:
            count += 1

    return count


library = []

add_book(
    library,
    1,
    "Python Crash Course",
    "Eric Matthes"
)

add_book(
    library,
    2,
    "Clean Code",
    "Robert Martin"
)

add_book(
    library,
    3,
    "Introduction to Algorithms",
    "Thomas Cormen"
)

add_book(
    library,
    4,
    "Deep Learning",
    "Ian Goodfellow"
)


show_library(library)

borrow_book(
    library,
    1,
    "Sumaiya"
)

borrow_book(
    library,
    3,
    "Rahim"
)

print(
    "\nAvailable books:",
    count_available_books(library)
)

print("\nSearch result:")

results = search_by_title(
    library,
    "Python"
)

for book in results:
    print(
        book["title"],
        "-",
        book["author"]
    )

return_book(library, 1)

show_library(library)
```

```python
menu = {
    "burger": 250,
    "pizza": 450,
    "pasta": 300,
    "fried chicken": 220,
    "coffee": 120,
    "juice": 150
}


def show_menu():

    print("\n========== MENU ==========")

    for item, price in menu.items():
        print(
            f"{item.title():20} Tk {price}"
        )


def add_to_order(order, item, quantity):

    item = item.lower()

    if item not in menu:
        print("Item is not available.")
        return

    if quantity <= 0:
        print("Quantity must be positive.")
        return

    if item in order:
        order[item] += quantity
    else:
        order[item] = quantity

    print(
        f"{quantity} {item} added."
    )


def remove_from_order(order, item):

    item = item.lower()

    if item not in order:
        print("Item is not in your order.")
        return

    del order[item]

    print(
        f"{item} removed from order."
    )


def calculate_total(order):

    total = 0

    for item, quantity in order.items():
        total += menu[item] * quantity

    return total


def calculate_discount(total):

    if total >= 5000:
        return total * 0.20

    if total >= 3000:
        return total * 0.15

    if total >= 1500:
        return total * 0.10

    return 0


def print_bill(order):

    if not order:
        print("Your order is empty.")
        return

    print("\n========== BILL ==========")

    for item, quantity in order.items():

        price = menu[item]
        item_total = price * quantity

        print(
            f"{item.title():20}"
            f"{quantity:3} x {price:4}"
            f" = {item_total}"
        )

    subtotal = calculate_total(order)

    discount = calculate_discount(
        subtotal
    )

    final_amount = subtotal - discount

    print("---------------------------")
    print("Subtotal :", subtotal)
    print("Discount :", discount)
    print("Total    :", final_amount)


def main():

    order = {}

    show_menu()

    add_to_order(
        order,
        "Pizza",
        2
    )

    add_to_order(
        order,
        "Burger",
        3
    )

    add_to_order(
        order,
        "Coffee",
        2
    )

    add_to_order(
        order,
        "Juice",
        1
    )

    remove_from_order(
        order,
        "coffee"
    )

    print_bill(order)


if __name__ == "__main__":
    main()
```

```python
def add_food(food_list, name, calories, meal):
    food = {
        "name": name,
        "calories": calories,
        "meal": meal
    }

    food_list.append(food)
    print(name, "added successfully.")


def show_foods(food_list):
    print("\n----- Today's Food -----")

    if not food_list:
        print("No food has been added yet.")
        return

    for food in food_list:
        print(
            f"{food['meal']}: "
            f"{food['name']} - "
            f"{food['calories']} calories"
        )


def total_calories(food_list):
    total = 0

    for food in food_list:
        total += food["calories"]

    return total


def meal_calories(food_list, meal):
    total = 0

    for food in food_list:
        if food["meal"].lower() == meal.lower():
            total += food["calories"]

    return total


def find_highest_calorie_food(food_list):
    if not food_list:
        return None

    highest = food_list[0]

    for food in food_list:
        if food["calories"] > highest["calories"]:
            highest = food

    return highest


def search_food(food_list, keyword):
    result = []

    for food in food_list:
        if keyword.lower() in food["name"].lower():
            result.append(food)

    return result


def calorie_status(food_list, daily_limit):
    total = total_calories(food_list)

    print("\n----- Calorie Status -----")
    print("Daily limit:", daily_limit)
    print("Consumed:", total)

    if total < daily_limit:
        remaining = daily_limit - total
        print("Remaining:", remaining, "calories")

    elif total == daily_limit:
        print("You reached your daily limit.")

    else:
        extra = total - daily_limit
        print("Exceeded by:", extra, "calories")


def daily_summary(food_list):
    total = total_calories(food_list)

    breakfast = meal_calories(
        food_list,
        "Breakfast"
    )

    lunch = meal_calories(
        food_list,
        "Lunch"
    )

    dinner = meal_calories(
        food_list,
        "Dinner"
    )

    snacks = meal_calories(
        food_list,
        "Snack"
    )

    print("\n========== DAILY SUMMARY ==========")

    print("Breakfast:", breakfast)
    print("Lunch:", lunch)
    print("Dinner:", dinner)
    print("Snacks:", snacks)
    print("Total:", total)


def compare_days(today, yesterday):
    today_total = total_calories(today)
    yesterday_total = total_calories(yesterday)

    print("\n----- Comparison -----")

    print("Yesterday:", yesterday_total)
    print("Today:", today_total)

    difference = today_total - yesterday_total

    if difference > 0:
        print(
            "Today you consumed",
            difference,
            "more calories."
        )

    elif difference < 0:
        print(
            "Today you consumed",
            abs(difference),
            "fewer calories."
        )

    else:
        print(
            "Both days have the same calories."
        )


def main():

    today = []

    yesterday = [
        {
            "name": "Rice",
            "calories": 300,
            "meal": "Lunch"
        },
        {
            "name": "Chicken",
            "calories": 250,
            "meal": "Lunch"
        },
        {
            "name": "Apple",
            "calories": 95,
            "meal": "Snack"
        }
    ]

    add_food(
        today,
        "Egg",
        140,
        "Breakfast"
    )

    add_food(
        today,
        "Bread",
        160,
        "Breakfast"
    )

    add_food(
        today,
        "Chicken Rice",
        450,
        "Lunch"
    )

    add_food(
        today,
        "Coffee",
        80,
        "Snack"
    )

    add_food(
        today,
        "Pizza",
        350,
        "Dinner"
    )

    show_foods(today)

    print(
        "\nTotal calories:",
        total_calories(today)
    )

    highest = find_highest_calorie_food(today)

    if highest:
        print(
            "\nHighest calorie food:",
            highest["name"]
        )

        print(
            "Calories:",
            highest["calories"]
        )

    print("\nSearch result for 'chicken':")

    result = search_food(
        today,
        "chicken"
    )

    if result:
        for food in result:
            print(
                food["name"],
                "-",
                food["calories"]
            )
    else:
        print("No food found.")

    calorie_status(
        today,
        1800
    )

    daily_summary(today)

    compare_days(
        today,
        yesterday
    )


main()
```

```python
from functools import reduce
from collections import Counter


# ==========================================
# MOVIE DATA
# ==========================================

movies = [
    {
        "title": "Inception",
        "genre": ["Sci-Fi", "Thriller"],
        "rating": 8.8,
        "year": 2010,
        "language": "English"
    },
    {
        "title": "Interstellar",
        "genre": ["Sci-Fi", "Drama"],
        "rating": 8.7,
        "year": 2014,
        "language": "English"
    },
    {
        "title": "The Dark Knight",
        "genre": ["Action", "Crime"],
        "rating": 9.0,
        "year": 2008,
        "language": "English"
    },
    {
        "title": "Parasite",
        "genre": ["Drama", "Thriller"],
        "rating": 8.5,
        "year": 2019,
        "language": "Korean"
    },
    {
        "title": "3 Idiots",
        "genre": ["Comedy", "Drama"],
        "rating": 8.4,
        "year": 2009,
        "language": "Hindi"
    },
    {
        "title": "Avengers: Endgame",
        "genre": ["Action", "Sci-Fi"],
        "rating": 8.4,
        "year": 2019,
        "language": "English"
    },
    {
        "title": "The Matrix",
        "genre": ["Sci-Fi", "Action"],
        "rating": 8.7,
        "year": 1999,
        "language": "English"
    }
]


# ==========================================
# BASIC FILTER FUNCTION
# ==========================================

def filter_by_rating(movie_list, minimum_rating):

    return [
        movie
        for movie in movie_list
        if movie["rating"] >= minimum_rating
    ]


# ==========================================
# GENRE SEARCH
# ==========================================

def find_by_genre(movie_list, genre):

    genre = genre.lower()

    return [
        movie
        for movie in movie_list
        if any(
            genre == g.lower()
            for g in movie["genre"]
        )
    ]


# ==========================================
# LANGUAGE FILTER
# ==========================================

def find_by_language(movie_list, language):

    return [
        movie
        for movie in movie_list
        if movie["language"].lower()
        == language.lower()
    ]


# ==========================================
# SORT MOVIES
# ==========================================

def sort_movies(movie_list, key_name):

    return sorted(
        movie_list,
        key=lambda movie: movie[key_name],
        reverse=True
    )


# ==========================================
# AVERAGE RATING
# ==========================================

def average_rating(movie_list):

    if not movie_list:
        return 0

    total = sum(
        movie["rating"]
        for movie in movie_list
    )

    return total / len(movie_list)


# ==========================================
# MOST COMMON GENRE
# ==========================================

def most_popular_genre(movie_list):

    all_genres = []

    for movie in movie_list:
        all_genres.extend(movie["genre"])

    if not all_genres:
        return None

    counter = Counter(all_genres)

    return counter.most_common(1)[0]


# ==========================================
# RECOMMENDATION SCORE
# ==========================================

def recommendation_score(movie, preferred_genres):

    score = movie["rating"]

    for genre in movie["genre"]:

        if genre.lower() in [
            g.lower()
            for g in preferred_genres
        ]:
            score += 1.5

    if movie["year"] >= 2015:
        score += 0.5

    return score


# ==========================================
# SMART RECOMMENDER
# ==========================================

def recommend_movies(
    movie_list,
    preferred_genres,
    minimum_rating=7.5
):

    candidates = filter_by_rating(
        movie_list,
        minimum_rating
    )

    ranked_movies = sorted(
        candidates,
        key=lambda movie:
        recommendation_score(
            movie,
            preferred_genres
        ),
        reverse=True
    )

    return ranked_movies


# ==========================================
# DISPLAY FUNCTION
# ==========================================

def display_movies(movie_list):

    if not movie_list:
        print("No movies found.")
        return

    print("\n" + "=" * 60)

    for index, movie in enumerate(
        movie_list,
        start=1
    ):

        genres = ", ".join(
            movie["genre"]
        )

        print(
            f"{index}. {movie['title']}"
        )

        print(
            f"   Rating   : {movie['rating']}"
        )

        print(
            f"   Genre    : {genres}"
        )

        print(
            f"   Year     : {movie['year']}"
        )

        print(
            f"   Language : {movie['language']}"
        )

    print("=" * 60)


# ==========================================
# FUNCTION COMPOSITION
# ==========================================

def create_movie_filter(
    minimum_rating,
    language=None
):

    def apply_filter(movie_list):

        result = filter_by_rating(
            movie_list,
            minimum_rating
        )

        if language:
            result = find_by_language(
                result,
                language
            )

        return result

    return apply_filter


# ==========================================
# MOVIE STATISTICS
# ==========================================

def movie_statistics(movie_list):

    print("\n========== MOVIE STATISTICS ==========")

    print(
        "Total Movies:",
        len(movie_list)
    )

    print(
        "Average Rating:",
        round(
            average_rating(movie_list),
            2
        )
    )

    popular = most_popular_genre(
        movie_list
    )

    if popular:
        print(
            "Popular Genre:",
            popular[0]
        )

        print(
            "Genre Count:",
            popular[1]
        )


# ==========================================
# MAIN PROGRAM
# ==========================================

print("\nALL MOVIES")
display_movies(movies)


# High-rated movies
high_rated = filter_by_rating(
    movies,
    8.5
)

print("\nHIGH RATED MOVIES")
display_movies(high_rated)


# Sci-Fi movies
sci_fi_movies = find_by_genre(
    movies,
    "Sci-Fi"
)

print("\nSCI-FI MOVIES")
display_movies(sci_fi_movies)


# Sort by rating
sorted_movies = sort_movies(
    movies,
    "rating"
)

print("\nMOVIES SORTED BY RATING")
display_movies(sorted_movies)


# Smart recommendations
preferences = [
    "Sci-Fi",
    "Action",
    "Thriller"
]

recommendations = recommend_movies(
    movies,
    preferences,
    minimum_rating=8.0
)

print("\nSMART RECOMMENDATIONS")
display_movies(recommendations)


# Closure-based filter
english_filter = create_movie_filter(
    minimum_rating=8.5,
    language="English"
)

english_movies = english_filter(
    movies
)

print(
    "\nHIGH-RATED ENGLISH MOVIES"
)

display_movies(
    english_movies
)


# Statistics
movie_statistics(movies)


# Final recommendation
if recommendations:

    best_movie = recommendations[0]

    print(
        "\nRecommended Movie:",
        best_movie["title"]
    )

    print(
        "Recommendation Score:",
        round(
            recommendation_score(
                best_movie,
                preferences
            ),
            2
        )
    )
```
