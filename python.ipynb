```python
# ==============================
# Python Functions Example
# ==============================

# Function to display a welcome message
def welcome(name):
    print(f"\nHello, {name}! Welcome to the Function Demo.\n")


# Function to calculate factorial using recursion
def factorial(n):
    if n == 0 or n == 1:
        return 1
    return n * factorial(n - 1)


# Function to check if a number is prime
def is_prime(num):
    if num < 2:
        return False
    for i in range(2, int(num ** 0.5) + 1):
        if num % i == 0:
            return False
    return True


# Function to calculate average
def average(numbers):
    return sum(numbers) / len(numbers)


# Main function
def main():
    name = input("Enter your name: ")
    welcome(name)

    number = int(input("Enter a number: "))

    print(f"\nFactorial of {number} = {factorial(number)}")

    if is_prime(number):
        print(f"{number} is a Prime Number.")
    else:
        print(f"{number} is NOT a Prime Number.")

    values = list(map(int, input("\nEnter numbers separated by spaces: ").split()))

    print(f"Average = {average(values):.2f}")


# Program starts here
if __name__ == "__main__":
    main()
```

```python
def fibonacci():
    cache = {}

    def fib(n):
        if n in cache:
            return cache[n]

        if n <= 1:
            return n

        cache[n] = fib(n - 1) + fib(n - 2)
        return cache[n]

    return fib

fib = fibonacci()

for i in range(20):
    print(fib(i), end=" ")
```

```python
import re

def validate_password(password):
    rules = {
        "Minimum Length": len(password) >= 8,
        "Uppercase": bool(re.search(r"[A-Z]", password)),
        "Lowercase": bool(re.search(r"[a-z]", password)),
        "Digit": bool(re.search(r"\d", password)),
        "Special Character": bool(re.search(r"[!@#$%^&*()_+=-]", password))
    }

    for rule, status in rules.items():
        print(f"{rule}: {'✔' if status else '✘'}")

    return all(rules.values())

password = input("Enter Password: ")

if validate_password(password):
    print("Strong Password")
else:
    print("Weak Password")
```

```python
import random
import time

def retry(attempts=3):
    def decorator(func):
        def wrapper(*args, **kwargs):
            for attempt in range(1, attempts + 1):
                try:
                    return func(*args, **kwargs)
                except Exception as e:
                    print(f"Attempt {attempt} Failed: {e}")
                    time.sleep(1)

            print("All attempts failed.")
        return wrapper
    return decorator

@retry(5)
def unstable():
    if random.randint(1, 4) != 1:
        raise Exception("Random Error")
    return "Success"

print(unstable())
```

```python
def word_frequency(sentence):

    words = sentence.lower().split()

    frequency = {}

    for word in words:
        frequency[word] = frequency.get(word, 0) + 1

    return frequency


text = input("Enter a sentence: ")

result = word_frequency(text)

print("\nWord Frequency:")

for word, count in result.items():
    print(word, ":", count)
```

```python
def encrypt(text, shift):
    encrypted = ""

    for char in text:
        if char.isalpha():
            start = 65 if char.isupper() else 97
            encrypted += chr((ord(char) - start + shift) % 26 + start)
        else:
            encrypted += char

    return encrypted


message = input("Enter Message: ")
shift = int(input("Enter Shift Value: "))

print("Encrypted Message:", encrypt(message, shift))
```

```python
def is_palindrome(sentence):
    cleaned = ""

    for char in sentence.lower():
        if char.isalnum():
            cleaned += char

    return cleaned == cleaned[::-1]


text = input("Enter a sentence: ")

if is_palindrome(text):
    print("Palindrome")
else:
    print("Not Palindrome")
```

```python
def password_score(password):

    score = 0

    if len(password) >= 8:
        score += 1

    if any(c.isupper() for c in password):
        score += 1

    if any(c.islower() for c in password):
        score += 1

    if any(c.isdigit() for c in password):
        score += 1

    if any(not c.isalnum() for c in password):
        score += 1

    levels = ["Very Weak", "Weak", "Average", "Good", "Strong", "Excellent"]

    return levels[score]


password = input("Enter Password: ")

print("Strength:", password_score(password))
```

```python
expenses = []


def add_expense():
    category = input("Enter Category: ")
    amount = float(input("Enter Amount: "))

    expenses.append({
        "Category": category,
        "Amount": amount
    })

    print("Expense Added Successfully!\n")


def view_expenses():
    if not expenses:
        print("No Expenses Found!\n")
        return

    print("\n----- Expense List -----")
    total = 0

    for i, expense in enumerate(expenses, start=1):
        print(f"{i}. {expense['Category']} - ${expense['Amount']:.2f}")
        total += expense["Amount"]

    print(f"\nTotal Expense: ${total:.2f}\n")


def highest_expense():
    if not expenses:
        print("No Expenses Found!\n")
        return

    highest = max(expenses, key=lambda x: x["Amount"])

    print("\nHighest Expense")
    print("Category:", highest["Category"])
    print("Amount: $", highest["Amount"])


def search_category():
    category = input("Enter Category to Search: ").lower()

    found = False

    for expense in expenses:
        if expense["Category"].lower() == category:
            print(f"{expense['Category']} - ${expense['Amount']:.2f}")
            found = True

    if not found:
        print("Category Not Found.")


def menu():
    while True:
        print("\n====== Expense Tracker ======")
        print("1. Add Expense")
        print("2. View Expenses")
        print("3. Highest Expense")
        print("4. Search by Category")
        print("5. Exit")

        choice = input("Enter Choice: ")

        if choice == "1":
            add_expense()

        elif choice == "2":
            view_expenses()

        elif choice == "3":
            highest_expense()

        elif choice == "4":
            search_category()

        elif choice == "5":
            print("Goodbye!")
            break

        else:
            print("Invalid Choice!")


menu()
```

```python
books = []


def add_book():
    title = input("Enter Book Title: ")
    author = input("Enter Author Name: ")

    books.append({
        "Title": title,
        "Author": author,
        "Available": True
    })

    print("Book Added Successfully!\n")


def view_books():
    if not books:
        print("No Books Available.\n")
        return

    print("\n----- Library Books -----")

    for i, book in enumerate(books, start=1):
        status = "Available" if book["Available"] else "Borrowed"
        print(f"{i}. {book['Title']} | {book['Author']} | {status}")


def borrow_book():
    title = input("Enter Book Title: ")

    for book in books:
        if book["Title"].lower() == title.lower():
            if book["Available"]:
                book["Available"] = False
                print("Book Borrowed Successfully!\n")
            else:
                print("Book is Already Borrowed.\n")
            return

    print("Book Not Found.\n")


def return_book():
    title = input("Enter Book Title: ")

    for book in books:
        if book["Title"].lower() == title.lower():
            if not book["Available"]:
                book["Available"] = True
                print("Book Returned Successfully!\n")
            else:
                print("Book is Already Available.\n")
            return

    print("Book Not Found.\n")


def menu():
    while True:
        print("\n====== Library Management ======")
        print("1. Add Book")
        print("2. View Books")
        print("3. Borrow Book")
        print("4. Return Book")
        print("5. Exit")

        choice = input("Enter Choice: ")

        if choice == "1":
            add_book()
        elif choice == "2":
            view_books()
        elif choice == "3":
            borrow_book()
        elif choice == "4":
            return_book()
        elif choice == "5":
            print("Thank You!")
            break
        else:
            print("Invalid Choice!\n")


menu()
```

```python
import time

def execution_time(func):
    def wrapper(*args, **kwargs):
        start = time.perf_counter()
        result = func(*args, **kwargs)
        end = time.perf_counter()

        print(f"\nFunction: {func.__name__}")
        print(f"Execution Time: {end - start:.6f} seconds")
        return result

    return wrapper


@execution_time
def find_prime_numbers(limit):
    primes = []

    for num in range(2, limit + 1):
        is_prime = True

        for i in range(2, int(num ** 0.5) + 1):
            if num % i == 0:
                is_prime = False
                break

        if is_prime:
            primes.append(num)

    return primes


@execution_time
def calculate_sum(limit):
    total = sum(range(1, limit + 1))
    return total


prime_list = find_prime_numbers(5000)
print(f"Total Prime Numbers: {len(prime_list)}")

total = calculate_sum(1_000_000)
print(f"Sum = {total}")
```

```python
import random
import time


def retry(max_attempts=3, delay=1):
    def decorator(func):
        def wrapper(*args, **kwargs):
            for attempt in range(1, max_attempts + 1):
                try:
                    result = func(*args, **kwargs)
                    print(f"Success on attempt {attempt}")
                    return result
                except Exception as error:
                    print(f"Attempt {attempt} failed: {error}")

                    if attempt < max_attempts:
                        print(f"Retrying in {delay} second(s)...\n")
                        time.sleep(delay)
                    else:
                        print("Maximum retry attempts reached.")
                        return None

        return wrapper
    return decorator


@retry(max_attempts=5, delay=2)
def connect_to_server():
    if random.choice([True, False]):
        raise ConnectionError("Server is unreachable!")

    return "Connected Successfully."


result = connect_to_server()

if result:
    print(result)
```

```python
def memoize(func):
    cache = {}

    def wrapper(*args):
        if args in cache:
            print(f"Cache Hit -> {args}")
            return cache[args]

        print(f"Calculating -> {args}")
        result = func(*args)
        cache[args] = result
        return result

    return wrapper


@memoize
def fibonacci(n):
    if n <= 1:
        return n
    return fibonacci(n - 1) + fibonacci(n - 2)


number = 35

print(f"Fibonacci({number}) =", fibonacci(number))
print(f"Fibonacci({number}) =", fibonacci(number))
```

```python
def scheduler():
    tasks = []

    def add_task(name, priority):
        tasks.append((priority, name))
        tasks.sort(reverse=True)

    def show_tasks():
        print("\nTask List")
        print("-" * 25)

        if not tasks:
            print("No tasks available.")
            return

        for priority, name in tasks:
            print(f"Priority {priority} -> {name}")

    def complete_task():
        if tasks:
            priority, name = tasks.pop(0)
            print(f"Completed: {name}")
        else:
            print("No task to complete.")

    return add_task, show_tasks, complete_task


add, show, complete = scheduler()

add("Finish Python Assignment", 2)
add("Prepare Presentation", 5)
add("Practice DSA", 4)
add("Upload GitHub Project", 3)

show()

complete()
complete()

show()
```

```python
def pipeline(*functions):

    def execute(data):
        for func in functions:
            data = func(data)
        return data

    return execute


def remove_spaces(text):
    return text.strip()


def to_upper(text):
    return text.upper()


def replace_spaces(text):
    return text.replace(" ", "_")


def add_extension(text):
    return text + ".txt"


process = pipeline(
    remove_spaces,
    to_upper,
    replace_spaces,
    add_extension
)

filename = process("   python function project   ")

print(filename)
```

```python
from datetime import datetime


def call_logger(func):
    history = []

    def wrapper(*args, **kwargs):
        result = func(*args, **kwargs)

        history.append({
            "time": datetime.now().strftime("%H:%M:%S"),
            "arguments": args,
            "result": result
        })

        return result

    wrapper.history = history
    return wrapper


@call_logger
def cube(number):
    return number ** 3


cube(2)
cube(5)
cube(10)

print("\nCall History")
print("-" * 35)

for item in cube.history:
    print(item)
```

```python
def count_calls(func):
    def wrapper(*args, **kwargs):
        wrapper.calls += 1
        print(f"{func.__name__} called {wrapper.calls} time(s)")
        return func(*args, **kwargs)

    wrapper.calls = 0
    return wrapper


@count_calls
def greet(name):
    print(f"Hello, {name}!")


greet("Alice")
greet("Bob")
greet("Charlie")
greet("David")

print("Total Calls:", greet.calls)
```

```python
def text_editor():
    text = ""
    history = []

    def write(new_text):
        nonlocal text
        history.append(text)
        text += new_text

    def undo():
        nonlocal text
        if history:
            text = history.pop()

    def display():
        print("Current Text:", text)

    return write, undo, display


write, undo, display = text_editor()

write("Python ")
write("Functions ")
write("are ")
write("Powerful!")

display()

undo()
display()

undo()
display()
```

```python
def transaction_manager(initial_balance):
    balance = initial_balance
    transactions = []

    def deposit(amount):
        nonlocal balance
        if amount > 0:
            balance += amount
            transactions.append(f"Deposited: {amount}")

    def withdraw(amount):
        nonlocal balance
        if amount <= balance:
            balance -= amount
            transactions.append(f"Withdrawn: {amount}")
        else:
            transactions.append("Withdrawal Failed")

    def statement():
        print("\nTransaction History")
        print("-" * 30)

        for item in transactions:
            print(item)

        print(f"\nCurrent Balance: {balance}")

    return deposit, withdraw, statement


deposit, withdraw, statement = transaction_manager(10000)

deposit(2500)
withdraw(1800)
withdraw(15000)
deposit(700)

statement()
```

```python
def add_book(library):
    print("\nAdd New Book")

    book_id = input("Book ID: ")

    for book in library:
        if book["id"] == book_id:
            print("Book ID already exists.")
            return

    title = input("Book Title: ")
    author = input("Author Name: ")

    while True:
        try:
            quantity = int(input("Quantity: "))
            if quantity >= 0:
                break
            else:
                print("Quantity cannot be negative.")
        except ValueError:
            print("Enter a valid number.")

    library.append({
        "id": book_id,
        "title": title,
        "author": author,
        "quantity": quantity
    })

    print("Book added successfully.\n")


def view_books(library):

    if len(library) == 0:
        print("\nNo books available.\n")
        return

    print("\nAvailable Books")
    print("-" * 60)

    for book in library:
        print(f"ID       : {book['id']}")
        print(f"Title    : {book['title']}")
        print(f"Author   : {book['author']}")
        print(f"Quantity : {book['quantity']}")
        print("-" * 60)


def search_book(library):

    keyword = input("Enter book title: ").lower()

    found = False

    for book in library:
        if keyword in book["title"].lower():
            print("\nBook Found")
            print(book)
            found = True

    if not found:
        print("Book not found.")


def borrow_book(library):

    book_id = input("Enter Book ID: ")

    for book in library:
        if book["id"] == book_id:

            if book["quantity"] > 0:
                book["quantity"] -= 1
                print("Book borrowed successfully.")
            else:
                print("Book is out of stock.")
            return

    print("Book not found.")


def return_book(library):

    book_id = input("Enter Book ID: ")

    for book in library:
        if book["id"] == book_id:
            book["quantity"] += 1
            print("Book returned successfully.")
            return

    print("Book not found.")


def update_book(library):

    book_id = input("Enter Book ID: ")

    for book in library:

        if book["id"] == book_id:

            print("\nLeave blank if you don't want to change.")

            new_title = input("New Title: ")
            new_author = input("New Author: ")
            new_quantity = input("New Quantity: ")

            if new_title != "":
                book["title"] = new_title

            if new_author != "":
                book["author"] = new_author

            if new_quantity != "":
                book["quantity"] = int(new_quantity)

            print("Book updated successfully.")
            return

    print("Book not found.")


def remove_book(library):

    book_id = input("Enter Book ID: ")

    for book in library:

        if book["id"] == book_id:
            library.remove(book)
            print("Book removed successfully.")
            return

    print("Book not found.")


def total_books(library):

    total = 0

    for book in library:
        total += book["quantity"]

    print("\nTotal Different Books :", len(library))
    print("Total Available Copies :", total)


def menu():

    library = []

    while True:

        print("""
========== Library Management ==========
1. Add Book
2. View Books
3. Search Book
4. Borrow Book
5. Return Book
6. Update Book
7. Remove Book
8. Total Books
9. Exit
========================================
""")

        choice = input("Enter your choice: ")

        if choice == "1":
            add_book(library)

        elif choice == "2":
            view_books(library)

        elif choice == "3":
            search_book(library)

        elif choice == "4":
            borrow_book(library)

        elif choice == "5":
            return_book(library)

        elif choice == "6":
            update_book(library)

        elif choice == "7":
            remove_book(library)

        elif choice == "8":
            total_books(library)

        elif choice == "9":
            print("Thank you for using the Library Management System.")
            break

        else:
            print("Invalid choice. Try again.")


def main():
    menu()


if __name__ == "__main__":
    main()
```
